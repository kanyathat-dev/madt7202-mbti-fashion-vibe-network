## Graph Construction and Reported Structure

- **Node set A:** 16 MBTI categories.
- **Node set B:** 12 Fashion Vibe categories.
- **Total nodes:** 28.
- **Possible cross-set links:** 16 × 12 = 192.
- **Edge weight:** image-level NPMI.
- **NPMI range:** −1 to 1.
- **Pruning rule:** retain observed edges with **NPMI > 0**.
- **Reported retained graph:** 51 edges out of 192 possible cross-set links.
- **Shortest-path distance:** **d = 1/NPMI** for retained positive-weight edges.

NPMI is an association weight; the inverse transformation is used as a distance for shortest-path analysis. Positive NPMI does not establish causality or statistical significance. A missing edge should not be interpreted as proof that a preference does not exist outside this sample.

- **Bridge load:** the project-reported quantity associated with no-overlap MBTI pairs. Its precise mathematical definition and normalization should be interpreted according to the executed code.
- **Shortest-path mediation:** an indirect route connecting MBTI categories that do not share a direct Vibe.
- **Leave-one-creator-out sensitivity:** compares bridge rankings after excluding each creator in turn.

```python
# =============================================================================
# MBTI x FASHION VIBE BIPARTITE NETWORK  -  MADT7202 Mid-Term
# Complete End-to-End Pipeline: Raw Data (67 Images / 268 Rows) -> Audit CSV
# -> NPMI Network -> Centrality + Pair Coverage -> Robustness Validation
# -> Figure 1 (ESTJ & INFP Shortest Path Highlight) -> Interactive HTML Suite
# =============================================================================
import itertools
from collections import Counter
import json

import numpy as np
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt
import IPython.display as display
from IPython.display import clear_output
import ipywidgets as widgets

# -----------------------------------------------------------------------------
# 1. RAW CODED DATA (Gemini output, 1 row = 1 attribute, 4 attributes/image)
#    Anonymous Creator mapping: Creator A-F (In compliance with PDPA & TikTok ToS)
# -----------------------------------------------------------------------------
mbti_ref = {
    "INTJ": ("Have original minds and great drive for implementing their ideas and achieving their goals. Skeptical and independent, have high standards of competence and performance.", "Original, Independent, Skeptical, High-Standards, Competence"),
    "ENTJ": ("Frank, decisive, assume leadership readily. Quickly see illogical and inefficient procedures, develop and implement comprehensive systems to solve problems.", "Decisive, Leadership, System-builder, Efficient, Forceful"),
    "ISTJ": ("Quiet, serious, earn success by thoroughness and dependability. Practical, matter-of-fact, realistic, and responsible. Decide logically what should be done.", "Dependable, Thorough, Realistic, Responsible, Orderly"),
    "ESTJ": ("Practical, realistic, matter-of-fact. Decisive, quickly move to implement decisions. Organize projects and people to get things done, focus on getting results in the most efficient way possible.", "Organized, Practical, Decisive, Results-driven, Structured"),
    "INTP": ("Seek to develop logical explanations for everything that interests them. Theoretical and abstract, interested more in ideas than in social interaction. Quiet, contained, flexible, and adaptable.", "Theoretical, Logical, Skeptical, Independent, Adaptable"),
    "ENTP": ("Quick, ingenious, stimulating, alert, and outspoken. Resourceful in solving new and challenging problems. Adept at generating conceptual possibilities and then analyzing them strategically.", "Inventive, Resourceful, Outspoken, Strategic, Non-conformist"),
    "ISTP": ("Tolerant and flexible, quiet observers until a problem appears, then act quickly to find workable solutions. Analyze what makes things work and readily get through large amounts of data to isolate core problems.", "Practical, Action-oriented, Observant, Logical, Resourceful"),
    "ESTP": ("Flexible and tolerant, take a pragmatic approach focused on immediate results. Bored by conceptual explanations; solve problems actively. Enjoy material comforts and style.", "Pragmatic, Spontaneous, Action-first, Energetic, Experiential"),
    "INFJ": ("Seek meaning and connection in ideas, relationships, and material possessions. Want to understand what motivates people and are insightful about others. Conscientious and committed to their firm values.", "Insightful, Meaning-seeking, Visionary, Conscientious, Holistic"),
    "INFP": ("Idealistic, loyal to their values and to people who are important to them. Want an external life that is congruent with their values. Curious, quick to see possibilities, can be catalysts for implementing ideas.", "Idealistic, Value-driven, Empathetic, Imaginative, Authentic"),
    "ISFJ": ("Quiet, friendly, responsible, and conscientious. Committed and steady in meeting their obligations. Thorough, painstaking, and accurate. Loyal, considerate, notice and remember specifics about people who are important to them.", "Warm, Conscientious, Loyal, Patient, Harmonious"),
    "ISFP": ("Quiet, friendly, sensitive, and kind. Enjoy the present moment, what's going on around them. Like to have their own space and to work within their own time frame. Loyal and committed to their values.", "Artistic, Sensitive, Non-judgmental, Spontaneous, Authentic"),
    "ESFJ": ("Warmhearted, conscientious, and cooperative. Want harmony in their environment, work with determination to establish it. Like to work with others to complete tasks accurately and on time. Loyal, follow through.", "Warm, Cooperative, Sociable, Harmonious, Loyal"),
    "ENFJ": ("Warm, empathetic, responsive, and responsible. Highly attuned to the emotions, needs, and motivations of others. Find potential in everyone, want to help others fulfill their potential. May act as catalysts for individual and group growth.", "Empathetic, Inspiring, Responsible, Collaborative, Expressive"),
    "ESFP": ("Outgoing, friendly, and accepting. Exuberant lovers of life, people, and material comforts. Enjoy working with others to make things happen. Bring common sense and a realistic approach to their work.", "Exuberant, Outgoing, Spontaneous, Playful, Generous"),
    "ENFP": ("Warmly enthusiastic and imaginative. See life as full of possibilities. Make connections between events and information very quickly, and confidently proceed based on the patterns they see. Want a lot of affirmation from others, readily give appreciation and support.", "Enthusiastic, Creative, Spontaneous, Empathetic, Visionary"),
}

raw_look_data = [
    # INTJ
    ("IMG_INTJ_001_L1", "INTJ", "Creator_A", "Sharp-Tailored", "tailored", "Black structured blazer with sharp shoulders", "Modern Executive / Minimal Luxury"),
    ("IMG_INTJ_001_L2", "INTJ", "Creator_A", "Sharp-Tailored", "formal", "Button-down dark shirt fully buttoned", "Modern Executive / Minimal Luxury"),
    ("IMG_INTJ_001_L3", "INTJ", "Creator_A", "Sharp-Tailored", "tailored", "Dark grey vest over a fitted long sleeve", "Modern Executive / Minimal Luxury"),
    ("IMG_INTJ_001_L4", "INTJ", "Creator_A", "Sharp-Tailored", "leather", "Black leather loafers and leather belt", "Modern Executive / Minimal Luxury"),
    ("IMG_INTJ_002_L1", "INTJ", "Creator_B", "Dark-Perfectionist", "dark_palette", "Black trench coat over dark grey top", "Dark Academia / Modern Gothic"),
    ("IMG_INTJ_002_L2", "INTJ", "Creator_B", "Dark-Perfectionist", "leather", "Knee-high black boots", "Dark Academia / Modern Gothic"),
    ("IMG_INTJ_002_L3", "INTJ", "Creator_B", "Dark-Perfectionist", "structured", "High-waisted tailored black trousers", "Dark Academia / Modern Gothic"),
    ("IMG_INTJ_002_L4", "INTJ", "Creator_B", "Dark-Perfectionist", "layered", "Black turtleneck under a dark grey blazer", "Dark Academia / Modern Gothic"),
    ("IMG_INTJ_003_L1", "INTJ", "Creator_C", "Clean-Traditional", "neutral_color", "Navy blue polo shirt with beige chinos", "Classic Contemporary / Hamptons Style"),
    ("IMG_INTJ_003_L2", "INTJ", "Creator_C", "Clean-Traditional", "formal", "Light blue button-down shirt tucked in", "Classic Contemporary / Hamptons Style"),
    ("IMG_INTJ_003_L3", "INTJ", "Creator_C", "Clean-Traditional", "knitwear", "Navy cardigan over a white t-shirt", "Classic Contemporary / Hamptons Style"),
    ("IMG_INTJ_003_L4", "INTJ", "Creator_C", "Clean-Traditional", "layered", "Trench coat in camel color", "Classic Contemporary / Hamptons Style"),
    ("IMG_INTJ_005_L1", "INTJ", "Creator_D", "Dark-Perfectionist", "dark_palette", "Dark plum and black color combination", "Dark Academia / Modern Gothic"),
    ("IMG_INTJ_005_L2", "INTJ", "Creator_D", "Dark-Perfectionist", "leather", "Pointed-toe black leather boots", "Dark Academia / Modern Gothic"),
    ("IMG_INTJ_005_L3", "INTJ", "Creator_D", "Dark-Perfectionist", "tailored", "Dark velvet blazer", "Dark Academia / Modern Gothic"),
    ("IMG_INTJ_005_L4", "INTJ", "Creator_D", "Dark-Perfectionist", "clean_silhouette", "Floor-length black tailored trousers", "Dark Academia / Modern Gothic"),
    # ENTJ
    ("IMG_ENTJ_001_L1", "ENTJ", "Creator_C", "Sharp-Tailored", "tailored", "Wide-shoulder padded blazer", "Modern Executive / Minimal Luxury"),
    ("IMG_ENTJ_001_L2", "ENTJ", "Creator_C", "Sharp-Tailored", "structured", "Thick woven fabric maintaining shape", "Modern Executive / Minimal Luxury"),
    ("IMG_ENTJ_001_L3", "ENTJ", "Creator_C", "Sharp-Tailored", "leather", "Large structured leather tote bag", "Modern Executive / Minimal Luxury"),
    ("IMG_ENTJ_001_L4", "ENTJ", "Creator_C", "Sharp-Tailored", "clean_silhouette", "Fitted midi dress with square neckline", "Modern Executive / Minimal Luxury"),
    ("IMG_ENTJ_002_L1", "ENTJ", "Creator_A", "Dark-Perfectionist", "dark_palette", "Black satin blouse with dark trousers", "Dark Academia / Modern Gothic"),
    ("IMG_ENTJ_002_L2", "ENTJ", "Creator_A", "Dark-Perfectionist", "tailored", "Slim-fit black suit", "Dark Academia / Modern Gothic"),
    ("IMG_ENTJ_002_L3", "ENTJ", "Creator_A", "Dark-Perfectionist", "leather", "Leather gloves and black coat", "Dark Academia / Modern Gothic"),
    ("IMG_ENTJ_002_L4", "ENTJ", "Creator_A", "Dark-Perfectionist", "layered", "Dark vest over a black long-sleeve top", "Dark Academia / Modern Gothic"),
    ("IMG_ENTJ_003_L1", "ENTJ", "Creator_D", "Utilitarian", "utilitarian", "Khaki cargo pants with large side pockets", "Industrial Loft / Urban Utility"),
    ("IMG_ENTJ_003_L2", "ENTJ", "Creator_D", "Utilitarian", "leather", "Chunky combat boots", "Industrial Loft / Urban Utility"),
    ("IMG_ENTJ_003_L3", "ENTJ", "Creator_D", "Utilitarian", "layered", "Tactical vest worn over a t-shirt", "Industrial Loft / Urban Utility"),
    ("IMG_ENTJ_003_L4", "ENTJ", "Creator_D", "Utilitarian", "denim", "Stiff dark denim jacket", "Industrial Loft / Urban Utility"),
    ("IMG_ENTJ_004_L1", "ENTJ", "Creator_C", "Sharp-Tailored", "tailored", "Structured pinstripe executive blazer", "Modern Executive / Minimal Luxury"),
    ("IMG_ENTJ_004_L2", "ENTJ", "Creator_C", "Sharp-Tailored", "formal", "High-collared formal button-up shirt with French cuffs", "Modern Executive / Minimal Luxury"),
    ("IMG_ENTJ_004_L3", "ENTJ", "Creator_C", "Sharp-Tailored", "leather", "Pointed leather ankle boots with gold buckle accents", "Modern Executive / Minimal Luxury"),
    ("IMG_ENTJ_004_L4", "ENTJ", "Creator_C", "Sharp-Tailored", "clean_silhouette", "Tailored ankle-length cigarette trousers with sharp crease", "Modern Executive / Minimal Luxury"),
    ("IMG_ENTJ_005_L1", "ENTJ", "Creator_C", "Sharp-Tailored", "tailored", "Navy blazer with gold buttons", "Modern Executive / Minimal Luxury"),
    ("IMG_ENTJ_005_L2", "ENTJ", "Creator_C", "Sharp-Tailored", "clean_silhouette", "Tapered navy trousers with front crease", "Modern Executive / Minimal Luxury"),
    ("IMG_ENTJ_005_L3", "ENTJ", "Creator_C", "Sharp-Tailored", "structured", "Tweed jacket with defined shoulders", "Modern Executive / Minimal Luxury"),
    ("IMG_ENTJ_005_L4", "ENTJ", "Creator_C", "Sharp-Tailored", "minimalist", "Simple black turtleneck dress", "Modern Executive / Minimal Luxury"),
    # ISTJ
    ("IMG_ISTJ_001_L1", "ISTJ", "Creator_D", "Sharp-Tailored", "tailored", "Dark structured blazer and straight-cut trousers", "Modern Executive / Minimal Luxury"),
    ("IMG_ISTJ_001_L2", "ISTJ", "Creator_D", "Sharp-Tailored", "formal", "Crisp white button-down shirt tucked in", "Modern Executive / Minimal Luxury"),
    ("IMG_ISTJ_001_L3", "ISTJ", "Creator_D", "Sharp-Tailored", "monochrome", "All grey suit setup from top to bottom", "Modern Executive / Minimal Luxury"),
    ("IMG_ISTJ_001_L4", "ISTJ", "Creator_D", "Sharp-Tailored", "leather", "Black leather belt and polished leather shoes", "Modern Executive / Minimal Luxury"),
    ("IMG_ISTJ_002_L1", "ISTJ", "Creator_A", "Clean-Traditional", "neutral_color", "Navy blue polo shirt with beige chinos", "Classic Contemporary / Hamptons Style"),
    ("IMG_ISTJ_002_L2", "ISTJ", "Creator_A", "Clean-Traditional", "knitwear", "Grey v-neck sweater worn over a collared shirt", "Classic Contemporary / Hamptons Style"),
    ("IMG_ISTJ_002_L3", "ISTJ", "Creator_A", "Clean-Traditional", "layered", "Beige trench coat worn over a button-up shirt", "Classic Contemporary / Hamptons Style"),
    ("IMG_ISTJ_002_L4", "ISTJ", "Creator_A", "Clean-Traditional", "minimalist", "Simple white t-shirt and straight dark jeans", "Classic Contemporary / Hamptons Style"),
    ("IMG_ISTJ_003_L1", "ISTJ", "Creator_C", "Dark-Perfectionist", "dark_palette", "Black overcoat with dark charcoal slacks", "Dark Academia / Modern Gothic"),
    ("IMG_ISTJ_003_L2", "ISTJ", "Creator_C", "Dark-Perfectionist", "monochrome", "All black outfit including top and bottom", "Dark Academia / Modern Gothic"),
    ("IMG_ISTJ_003_L3", "ISTJ", "Creator_C", "Dark-Perfectionist", "formal", "Stiff dark collar and formal cuffs", "Dark Academia / Modern Gothic"),
    ("IMG_ISTJ_003_L4", "ISTJ", "Creator_C", "Dark-Perfectionist", "clean_silhouette", "Sleek dark turtleneck tucked into trousers", "Dark Academia / Modern Gothic"),
    ("IMG_ISTJ_005_L1", "ISTJ", "Creator_B", "Sharp-Tailored", "tailored", "Navy blue two-piece suit", "Modern Executive / Minimal Luxury"),
    ("IMG_ISTJ_005_L2", "ISTJ", "Creator_B", "Sharp-Tailored", "clean_silhouette", "Fitted dark pencil skirt", "Modern Executive / Minimal Luxury"),
    ("IMG_ISTJ_005_L3", "ISTJ", "Creator_B", "Sharp-Tailored", "structured", "Double-breasted coat with sharp lapels", "Modern Executive / Minimal Luxury"),
    ("IMG_ISTJ_005_L4", "ISTJ", "Creator_B", "Sharp-Tailored", "monochrome", "Black blazer paired with black trousers", "Modern Executive / Minimal Luxury"),
    # ESTJ
    ("IMG_ESTJ_001_L1", "ESTJ", "Creator_C", "Sharp-Tailored", "tailored", "Wide-shoulder padded blazer", "Modern Executive / Minimal Luxury"),
    ("IMG_ESTJ_001_L2", "ESTJ", "Creator_C", "Sharp-Tailored", "structured", "Stiff fabric skirt suit maintaining its shape", "Modern Executive / Minimal Luxury"),
    ("IMG_ESTJ_001_L3", "ESTJ", "Creator_C", "Sharp-Tailored", "monochrome", "All white tailored suit set", "Modern Executive / Minimal Luxury"),
    ("IMG_ESTJ_001_L4", "ESTJ", "Creator_C", "Sharp-Tailored", "formal", "Classic pointed-toe heels", "Modern Executive / Minimal Luxury"),
    ("IMG_ESTJ_002_L1", "ESTJ", "Creator_A", "Clean-Traditional", "neutral_color", "Camel-colored cardigan over a white blouse", "Classic Contemporary / Hamptons Style"),
    ("IMG_ESTJ_002_L2", "ESTJ", "Creator_A", "Clean-Traditional", "tailored", "Crisp polo shirt tucked into khaki trousers", "Classic Contemporary / Hamptons Style"),
    ("IMG_ESTJ_002_L3", "ESTJ", "Creator_A", "Clean-Traditional", "layered", "Navy blazer worn over a light blue button-down", "Classic Contemporary / Hamptons Style"),
    ("IMG_ESTJ_002_L4", "ESTJ", "Creator_A", "Clean-Traditional", "leather", "Brown leather loafers and matching watch strap", "Classic Contemporary / Hamptons Style"),
    ("IMG_ESTJ_003_L1", "ESTJ", "Creator_B", "Sharp-Tailored", "tailored", "Charcoal grey suit with a fitted blazer", "Modern Executive / Minimal Luxury"),
    ("IMG_ESTJ_003_L2", "ESTJ", "Creator_B", "Sharp-Tailored", "formal", "Silk tie with a windsor knot", "Modern Executive / Minimal Luxury"),
    ("IMG_ESTJ_003_L3", "ESTJ", "Creator_B", "Sharp-Tailored", "structured", "Stiff collared shirt maintaining shape", "Modern Executive / Minimal Luxury"),
    ("IMG_ESTJ_003_L4", "ESTJ", "Creator_B", "Sharp-Tailored", "leather", "Polished black oxfords and leather briefcase", "Modern Executive / Minimal Luxury"),
    ("IMG_ESTJ_005_L1", "ESTJ", "Creator_D", "Sharp-Tailored", "tailored", "Pinstripe blazer perfectly fitted at shoulders", "Modern Executive / Minimal Luxury"),
    ("IMG_ESTJ_005_L2", "ESTJ", "Creator_D", "Sharp-Tailored", "structured", "Black double-breasted suit jacket", "Modern Executive / Minimal Luxury"),
    ("IMG_ESTJ_005_L3", "ESTJ", "Creator_D", "Sharp-Tailored", "clean_silhouette", "Sleek turtleneck worn under a tailored coat", "Modern Executive / Minimal Luxury"),
    ("IMG_ESTJ_005_L4", "ESTJ", "Creator_D", "Sharp-Tailored", "leather", "Pointed leather heels and leather belt", "Modern Executive / Minimal Luxury"),
    # INTP
    ("IMG_INTP_001_L1", "INTP", "Creator_D", "Normcore", "casual", "Basic grey t-shirt with no print", "Warm Scandinavian / Basic Japandi"),
    ("IMG_INTP_001_L2", "INTP", "Creator_D", "Normcore", "denim", "Straight fit medium-wash blue jeans", "Warm Scandinavian / Basic Japandi"),
    ("IMG_INTP_001_L3", "INTP", "Creator_D", "Normcore", "oversized", "Baggy dark zip-up hoodie", "Warm Scandinavian / Basic Japandi"),
    ("IMG_INTP_001_L4", "INTP", "Creator_D", "Normcore", "knitwear", "Plain black crewneck sweater", "Warm Scandinavian / Basic Japandi"),
    ("IMG_INTP_002_L1", "INTP", "Creator_A", "Clean-Traditional", "neutral_color", "Beige chinos and crisp white shirt", "Classic Contemporary / Hamptons Style"),
    ("IMG_INTP_002_L2", "INTP", "Creator_A", "Clean-Traditional", "knitwear", "Navy blue fine-knit cardigan", "Classic Contemporary / Hamptons Style"),
    ("IMG_INTP_002_L3", "INTP", "Creator_A", "Clean-Traditional", "layered", "Sweater worn neatly over a collared shirt", "Classic Contemporary / Hamptons Style"),
    ("IMG_INTP_002_L4", "INTP", "Creator_A", "Clean-Traditional", "tailored", "Well-fitted straight trousers", "Classic Contemporary / Hamptons Style"),
    ("IMG_INTP_003_L1", "INTP", "Creator_C", "Dark-Perfectionist", "monochrome", "All-black outfit from top to bottom", "Dark Academia / Modern Gothic"),
    ("IMG_INTP_003_L2", "INTP", "Creator_C", "Dark-Perfectionist", "clean_silhouette", "Straight sleek dark pants with no wrinkles", "Dark Academia / Modern Gothic"),
    ("IMG_INTP_003_L3", "INTP", "Creator_C", "Dark-Perfectionist", "tailored", "Structured black blazer", "Dark Academia / Modern Gothic"),
    ("IMG_INTP_003_L4", "INTP", "Creator_C", "Dark-Perfectionist", "leather", "Polished black leather boots", "Dark Academia / Modern Gothic"),
    ("IMG_INTP_005_L1", "INTP", "Creator_D", "Normcore", "casual", "Relaxed grey sweatpants", "Warm Scandinavian / Basic Japandi"),
    ("IMG_INTP_005_L2", "INTP", "Creator_D", "Normcore", "denim", "Relaxed fit light wash blue jeans", "Warm Scandinavian / Basic Japandi"),
    ("IMG_INTP_005_L3", "INTP", "Creator_D", "Normcore", "knitwear", "Chunky beige knit sweater", "Warm Scandinavian / Basic Japandi"),
    ("IMG_INTP_005_L4", "INTP", "Creator_D", "Normcore", "casual", "Plain canvas tote bag on the shoulder", "Warm Scandinavian / Basic Japandi"),
    # ENTP
    ("IMG_ENTP_001_L1", "ENTP", "Creator_C", "Deconstructed", "deconstructed", "Shirt with a distinct asymmetrical hemline", "Brutalism / Deconstructivism Interior"),
    ("IMG_ENTP_001_L2", "ENTP", "Creator_C", "Deconstructed", "distressed", "Ripped and frayed edges on the denim", "Brutalism / Deconstructivism Interior"),
    ("IMG_ENTP_001_L3", "ENTP", "Creator_C", "Deconstructed", "oversized", "Excessively long sleeves covering the hands", "Brutalism / Deconstructivism Interior"),
    ("IMG_ENTP_001_L4", "ENTP", "Creator_C", "Deconstructed", "streetwear", "Chunky sneakers with abstract irregular shapes", "Brutalism / Deconstructivism Interior"),
    ("IMG_ENTP_002_L1", "ENTP", "Creator_D", "Utilitarian", "utilitarian", "Nylon multi-pocket tactical vest", "Industrial Loft / Urban Utility"),
    ("IMG_ENTP_002_L2", "ENTP", "Creator_D", "Utilitarian", "leather", "Thick black leather combat boots", "Industrial Loft / Urban Utility"),
    ("IMG_ENTP_002_L3", "ENTP", "Creator_D", "Utilitarian", "layered", "Long sleeve shirt worn under a short sleeve button-up", "Industrial Loft / Urban Utility"),
    ("IMG_ENTP_002_L4", "ENTP", "Creator_D", "Utilitarian", "casual", "Simple black knit beanie", "Industrial Loft / Urban Utility"),
    ("IMG_ENTP_003_L1", "ENTP", "Creator_A", "Playful-Streetwear", "colorful", "Bright red graphic hoodie", "Postmodern / Memphis Design / Y2K Loft"),
    ("IMG_ENTP_003_L2", "ENTP", "Creator_A", "Playful-Streetwear", "bold_pattern", "Black and white checkered slip-on shoes", "Postmodern / Memphis Design / Y2K Loft"),
    ("IMG_ENTP_003_L3", "ENTP", "Creator_A", "Playful-Streetwear", "oversized", "Huge vintage hockey jersey", "Postmodern / Memphis Design / Y2K Loft"),
    ("IMG_ENTP_003_L4", "ENTP", "Creator_A", "Playful-Streetwear", "denim", "Paint-splattered baggy denim pants", "Postmodern / Memphis Design / Y2K Loft"),
    ("IMG_ENTP_005_L1", "ENTP", "Creator_D", "Playful-Streetwear", "colorful", "Neon green knitted beanie", "Postmodern / Memphis Design / Y2K Loft"),
    ("IMG_ENTP_005_L2", "ENTP", "Creator_D", "Playful-Streetwear", "denim", "Light wash baggy denim jorts (shorts)", "Postmodern / Memphis Design / Y2K Loft"),
    ("IMG_ENTP_005_L3", "ENTP", "Creator_D", "Playful-Streetwear", "oversized", "Very wide leg grey sweatpants", "Postmodern / Memphis Design / Y2K Loft"),
    ("IMG_ENTP_005_L4", "ENTP", "Creator_D", "Playful-Streetwear", "bold_pattern", "Striped long sleeve shirt worn under a t-shirt", "Postmodern / Memphis Design / Y2K Loft"),
    # ISTP
    ("IMG_ISTP_001_L1", "ISTP", "Creator_C", "Utilitarian", "utilitarian", "Black cargo pants with large side pockets", "Industrial Loft / Urban Utility"),
    ("IMG_ISTP_001_L2", "ISTP", "Creator_C", "Utilitarian", "leather", "Thick black leather combat boots", "Industrial Loft / Urban Utility"),
    ("IMG_ISTP_001_L3", "ISTP", "Creator_C", "Utilitarian", "layered", "Tactical vest worn over a dark long sleeve shirt", "Industrial Loft / Urban Utility"),
    ("IMG_ISTP_001_L4", "ISTP", "Creator_C", "Utilitarian", "denim", "Stiff dark washed denim overshirt", "Industrial Loft / Urban Utility"),
    ("IMG_ISTP_002_L1", "ISTP", "Creator_A", "Normcore", "casual", "Plain grey pullover hoodie with no logos", "Warm Scandinavian / Basic Japandi"),
    ("IMG_ISTP_002_L2", "ISTP", "Creator_A", "Normcore", "denim", "Standard straight fit blue jeans", "Warm Scandinavian / Basic Japandi"),
    ("IMG_ISTP_002_L3", "ISTP", "Creator_A", "Normcore", "casual", "Basic white canvas sneakers", "Warm Scandinavian / Basic Japandi"),
    ("IMG_ISTP_002_L4", "ISTP", "Creator_A", "Normcore", "knitwear", "Plain dark navy crewneck sweater", "Warm Scandinavian / Basic Japandi"),
    ("IMG_ISTP_003_L1", "ISTP", "Creator_D", "Deconstructed", "deconstructed", "Shirt with asymmetrical uneven hemline", "Brutalism / Deconstructivism Interior"),
    ("IMG_ISTP_003_L2", "ISTP", "Creator_D", "Deconstructed", "layered", "Multiple t-shirts layered unevenly", "Brutalism / Deconstructivism Interior"),
    ("IMG_ISTP_003_L3", "ISTP", "Creator_D", "Deconstructed", "oversized", "Jacket sleeves extending past the hands", "Brutalism / Deconstructivism Interior"),
    ("IMG_ISTP_003_L4", "ISTP", "Creator_D", "Deconstructed", "streetwear", "Chunky distressed sneakers", "Brutalism / Deconstructivism Interior"),
    ("IMG_ISTP_005_L1", "ISTP", "Creator_B", "Clean-Traditional", "minimalist", "Simple fitted white t-shirt", "Classic Contemporary / Hamptons Style"),
    ("IMG_ISTP_005_L2", "ISTP", "Creator_B", "Clean-Traditional", "leather", "Clean brown leather belt and matching boots", "Classic Contemporary / Hamptons Style"),
    ("IMG_ISTP_005_L3", "ISTP", "Creator_B", "Clean-Traditional", "formal", "Silver metal watch on the wrist", "Classic Contemporary / Hamptons Style"),
    ("IMG_ISTP_005_L4", "ISTP", "Creator_B", "Clean-Traditional", "casual", "Spotless white leather sneakers", "Classic Contemporary / Hamptons Style"),
    # ESTP
    ("IMG_ESTP_001_L1", "ESTP", "Creator_D", "High-Energy", "sporty", "Bright red track pants with side stripes", "High-Tech Dynamic / Neon Pop Modern"),
    ("IMG_ESTP_001_L2", "ESTP", "Creator_D", "High-Energy", "streetwear", "Oversized sports jersey with large numbers", "High-Tech Dynamic / Neon Pop Modern"),
    ("IMG_ESTP_001_L3", "ESTP", "Creator_D", "High-Energy", "layered", "Nylon windbreaker jacket in bright blue", "High-Tech Dynamic / Neon Pop Modern"),
    ("IMG_ESTP_001_L4", "ESTP", "Creator_D", "High-Energy", "edgy", "Reflective sporty sunglasses", "High-Tech Dynamic / Neon Pop Modern"),
    ("IMG_ESTP_002_L1", "ESTP", "Creator_C", "Playful-Streetwear", "streetwear", "Baggy graphic t-shirt with skate logos", "Postmodern / Memphis Design / Y2K Loft"),
    ("IMG_ESTP_002_L2", "ESTP", "Creator_C", "Playful-Streetwear", "colorful", "Bright orange beanie hat", "Postmodern / Memphis Design / Y2K Loft"),
    ("IMG_ESTP_002_L3", "ESTP", "Creator_C", "Playful-Streetwear", "layered", "Long sleeve striped shirt worn under a t-shirt", "Postmodern / Memphis Design / Y2K Loft"),
    ("IMG_ESTP_002_L4", "ESTP", "Creator_C", "Playful-Streetwear", "edgy", "Thick silver chain hanging from the belt loop", "Postmodern / Memphis Design / Y2K Loft"),
    ("IMG_ESTP_003_L1", "ESTP", "Creator_B", "Glam-Maximalist", "leather", "Shiny black leather biker jacket", "Hollywood Regency / Art Deco Maximalism"),
    ("IMG_ESTP_003_L2", "ESTP", "Creator_B", "Glam-Maximalist", "bold_pattern", "Leopard print shirt worn underneath the jacket", "Hollywood Regency / Art Deco Maximalism"),
    ("IMG_ESTP_003_L3", "ESTP", "Creator_B", "Glam-Maximalist", "formal", "Pointed toe stilettos in bright red", "Hollywood Regency / Art Deco Maximalism"),
    ("IMG_ESTP_003_L4", "ESTP", "Creator_B", "Glam-Maximalist", "layered", "Multiple chunky gold necklaces stacked", "Hollywood Regency / Art Deco Maximalism"),
    ("IMG_ESTP_005_L1", "ESTP", "Creator_A", "High-Energy", "sporty", "Tight black biker shorts", "High-Tech Dynamic / Neon Pop Modern"),
    ("IMG_ESTP_005_L2", "ESTP", "Creator_A", "High-Energy", "colorful", "Bright color contrast between pink and black", "High-Tech Dynamic / Neon Pop Modern"),
    ("IMG_ESTP_005_L3", "ESTP", "Creator_A", "High-Energy", "casual", "Nylon sporty crossbody bag", "High-Tech Dynamic / Neon Pop Modern"),
    ("IMG_ESTP_005_L4", "ESTP", "Creator_A", "High-Energy", "layered", "White ribbed tank top visible under hoodie", "High-Tech Dynamic / Neon Pop Modern"),
    # INFJ
    ("IMG_INFJ_001_L1", "INFJ", "Creator_A", "Dark-Perfectionist", "dark_palette", "Black turtleneck sweater", "Dark Academia / Modern Gothic"),
    ("IMG_INFJ_001_L2", "INFJ", "Creator_A", "Dark-Perfectionist", "tailored", "Long dark charcoal overcoat", "Dark Academia / Modern Gothic"),
    ("IMG_INFJ_001_L3", "INFJ", "Creator_A", "Dark-Perfectionist", "monochrome", "All-black outfit from head to toe", "Dark Academia / Modern Gothic"),
    ("IMG_INFJ_001_L4", "INFJ", "Creator_A", "Dark-Perfectionist", "leather", "Polished black leather boots", "Dark Academia / Modern Gothic"),
    ("IMG_INFJ_002_L1", "INFJ", "Creator_C", "Ethereal-Soft", "soft_fabric", "Flowing chiffon maxi skirt", "Soft Minimal / Cloud Zen"),
    ("IMG_INFJ_002_L2", "INFJ", "Creator_C", "Ethereal-Soft", "feminine", "Blouse with soft ruffled details", "Soft Minimal / Cloud Zen"),
    ("IMG_INFJ_002_L3", "INFJ", "Creator_C", "Ethereal-Soft", "layered", "Thin sheer top worn over a white camisole", "Soft Minimal / Cloud Zen"),
    ("IMG_INFJ_002_L4", "INFJ", "Creator_C", "Ethereal-Soft", "delicate", "Thin strappy sandals", "Soft Minimal / Cloud Zen"),
    ("IMG_INFJ_003_L1", "INFJ", "Creator_D", "Soft-Romantic", "romantic", "Lace trim around the collar", "French Cottage / Vintage Shabby Chic"),
    ("IMG_INFJ_003_L2", "INFJ", "Creator_D", "Soft-Romantic", "feminine", "Fitted bodice dress with flared skirt", "French Cottage / Vintage Shabby Chic"),
    ("IMG_INFJ_003_L3", "INFJ", "Creator_D", "Soft-Romantic", "vintage", "Blouse with puffed sleeves", "French Cottage / Vintage Shabby Chic"),
    ("IMG_INFJ_003_L4", "INFJ", "Creator_D", "Soft-Romantic", "elegant", "Classic black ballet flats", "French Cottage / Vintage Shabby Chic"),
    ("IMG_INFJ_005_L1", "INFJ", "Creator_D", "Earthy-Raw", "neutral_color", "Olive green canvas trousers", "Wabi-Sabi / Earthy Japandi"),
    ("IMG_INFJ_005_L2", "INFJ", "Creator_D", "Earthy-Raw", "earthy", "Brown corduroy jacket", "Wabi-Sabi / Earthy Japandi"),
    ("IMG_INFJ_005_L3", "INFJ", "Creator_D", "Earthy-Raw", "layered", "Knit vest worn over a long sleeve shirt", "Wabi-Sabi / Earthy Japandi"),
    ("IMG_INFJ_005_L4", "INFJ", "Creator_D", "Earthy-Raw", "relaxed", "Baggy fit trousers allowing movement", "Wabi-Sabi / Earthy Japandi"),
    # INFP
    ("IMG_INFP_001_L1", "INFP", "Creator_D", "Earthy-Raw", "neutral_color", "Beige overall dress", "Wabi-Sabi / Earthy Japandi"),
    ("IMG_INFP_001_L2", "INFP", "Creator_D", "Earthy-Raw", "linen", "Beige linen wide-leg pants", "Wabi-Sabi / Earthy Japandi"),
    ("IMG_INFP_001_L3", "INFP", "Creator_D", "Earthy-Raw", "layered", "Brown cardigan worn open over a white tee", "Wabi-Sabi / Earthy Japandi"),
    ("IMG_INFP_001_L4", "INFP", "Creator_D", "Earthy-Raw", "relaxed", "Loose smock dress with no defined waist", "Wabi-Sabi / Earthy Japandi"),
    ("IMG_INFP_002_L1", "INFP", "Creator_A", "Soft-Romantic", "romantic", "White peasant blouse with ruffled neckline", "French Cottage / Vintage Shabby Chic"),
    ("IMG_INFP_002_L2", "INFP", "Creator_A", "Soft-Romantic", "feminine", "Black Mary Jane shoes with low heels", "French Cottage / Vintage Shabby Chic"),
    ("IMG_INFP_002_L3", "INFP", "Creator_A", "Soft-Romantic", "soft_fabric", "Light breathable cotton dress", "French Cottage / Vintage Shabby Chic"),
    ("IMG_INFP_002_L4", "INFP", "Creator_A", "Soft-Romantic", "vintage", "Ribbon tied in a bow at the neck", "French Cottage / Vintage Shabby Chic"),
    ("IMG_INFP_003_L1", "INFP", "Creator_B", "Ethereal-Soft", "flowing", "Long white maxi skirt billowing outwards", "Soft Minimal / Cloud Zen"),
    ("IMG_INFP_003_L2", "INFP", "Creator_B", "Ethereal-Soft", "pastel", "Mint green silk top", "Soft Minimal / Cloud Zen"),
    ("IMG_INFP_003_L3", "INFP", "Creator_B", "Ethereal-Soft", "delicate", "Thin spaghetti strap camisole", "Soft Minimal / Cloud Zen"),
    ("IMG_INFP_003_L4", "INFP", "Creator_B", "Ethereal-Soft", "layered", "Lightweight mesh top worn over a bralette", "Soft Minimal / Cloud Zen"),
    ("IMG_INFP_005_L1", "INFP", "Creator_C", "Earthy-Raw", "earthy", "Rust orange corduroy pants", "Wabi-Sabi / Earthy Japandi"),
    ("IMG_INFP_005_L2", "INFP", "Creator_C", "Earthy-Raw", "vintage", "Retro structured leather satchel", "Wabi-Sabi / Earthy Japandi"),
    ("IMG_INFP_005_L3", "INFP", "Creator_C", "Earthy-Raw", "layered", "Plaid flannel shirt worn open over a t-shirt", "Wabi-Sabi / Earthy Japandi"),
    ("IMG_INFP_005_L4", "INFP", "Creator_C", "Earthy-Raw", "relaxed", "Straight leg loose brown denim", "Wabi-Sabi / Earthy Japandi"),
    # ISFJ
    ("IMG_ISFJ_001_L1", "ISFJ", "Creator_D", "Soft-Romantic", "romantic", "White blouse with delicate lace collar", "French Cottage / Vintage Shabby Chic"),
    ("IMG_ISFJ_001_L2", "ISFJ", "Creator_D", "Soft-Romantic", "feminine", "Pleated midi skirt flowing past the knees", "French Cottage / Vintage Shabby Chic"),
    ("IMG_ISFJ_001_L3", "ISFJ", "Creator_D", "Soft-Romantic", "floral", "Subtle small floral print on the skirt", "French Cottage / Vintage Shabby Chic"),
    ("IMG_ISFJ_001_L4", "ISFJ", "Creator_D", "Soft-Romantic", "elegant", "Classic beige ballet flats", "French Cottage / Vintage Shabby Chic"),
    ("IMG_ISFJ_002_L1", "ISFJ", "Creator_A", "Ethereal-Soft", "flowing", "Long white chiffon maxi dress", "Soft Minimal / Cloud Zen"),
    ("IMG_ISFJ_002_L2", "ISFJ", "Creator_A", "Ethereal-Soft", "soft_fabric", "Sheer fabric overlay on the dress", "Soft Minimal / Cloud Zen"),
    ("IMG_ISFJ_002_L3", "ISFJ", "Creator_A", "Ethereal-Soft", "feminine", "Soft wavy hair resting on the shoulders", "Soft Minimal / Cloud Zen"),
    ("IMG_ISFJ_002_L4", "ISFJ", "Creator_A", "Ethereal-Soft", "layered", "Lightweight silk shawl draped over shoulders", "Soft Minimal / Cloud Zen"),
    ("IMG_ISFJ_003_L1", "ISFJ", "Creator_C", "Clean-Traditional", "knitwear", "Navy blue crewneck sweater", "Classic Contemporary / Hamptons Style"),
    ("IMG_ISFJ_003_L2", "ISFJ", "Creator_C", "Clean-Traditional", "tailored", "Straight-cut beige trousers", "Classic Contemporary / Hamptons Style"),
    ("IMG_ISFJ_003_L3", "ISFJ", "Creator_C", "Clean-Traditional", "leather", "Brown leather loafers", "Classic Contemporary / Hamptons Style"),
    ("IMG_ISFJ_003_L4", "ISFJ", "Creator_C", "Clean-Traditional", "minimalist", "Clean outfit with no logos", "Classic Contemporary / Hamptons Style"),
    ("IMG_ISFJ_005_L1", "ISFJ", "Creator_E", "Ethereal-Soft", "soft_fabric", "Satin silk slip dress in champagne color", "Soft Minimal / Cloud Zen"),
    ("IMG_ISFJ_005_L2", "ISFJ", "Creator_E", "Ethereal-Soft", "layered", "Chunky oversized white knit cardigan over the dress", "Soft Minimal / Cloud Zen"),
    ("IMG_ISFJ_005_L3", "ISFJ", "Creator_E", "Ethereal-Soft", "feminine", "Hair tied back with a silk scrunchie", "Soft Minimal / Cloud Zen"),
    ("IMG_ISFJ_005_L4", "ISFJ", "Creator_E", "Ethereal-Soft", "casual", "Soft suede mules", "Soft Minimal / Cloud Zen"),
    # ISFP
    ("IMG_ISFP_001_L1", "ISFP", "Creator_D", "Earthy-Raw", "earthy", "Rust orange linen button-down shirt", "Wabi-Sabi / Earthy Japandi"),
    ("IMG_ISFP_001_L2", "ISFP", "Creator_D", "Earthy-Raw", "casual", "White canvas sneakers slightly scuffed", "Wabi-Sabi / Earthy Japandi"),
    ("IMG_ISFP_001_L3", "ISFP", "Creator_D", "Earthy-Raw", "linen", "Natural wrinkled texture on the shirt", "Wabi-Sabi / Earthy Japandi"),
    ("IMG_ISFP_001_L4", "ISFP", "Creator_D", "Earthy-Raw", "natural", "Woven canvas tote bag", "Wabi-Sabi / Earthy Japandi"),
    ("IMG_ISFP_002_L1", "ISFP", "Creator_A", "Soft-Romantic", "floral", "Flowing maxi skirt with large daisy prints", "French Cottage / Vintage Shabby Chic"),
    ("IMG_ISFP_002_L2", "ISFP", "Creator_A", "Soft-Romantic", "feminine", "Wavy hair worn down naturally", "French Cottage / Vintage Shabby Chic"),
    ("IMG_ISFP_002_L3", "ISFP", "Creator_A", "Soft-Romantic", "soft_fabric", "Light breathable cotton material", "French Cottage / Vintage Shabby Chic"),
    ("IMG_ISFP_002_L4", "ISFP", "Creator_A", "Soft-Romantic", "casual", "Simple brown leather sandals", "French Cottage / Vintage Shabby Chic"),
    ("IMG_ISFP_003_L1", "ISFP", "Creator_C", "Playful-Streetwear", "colorful", "Bright blue oversized graphic hoodie", "Postmodern / Memphis Design / Y2K Loft"),
    ("IMG_ISFP_003_L2", "ISFP", "Creator_C", "Playful-Streetwear", "casual", "Chunky skate sneakers with thick laces", "Postmodern / Memphis Design / Y2K Loft"),
    ("IMG_ISFP_003_L3", "ISFP", "Creator_C", "Playful-Streetwear", "denim", "Light wash denim with paint splatters", "Postmodern / Memphis Design / Y2K Loft"),
    ("IMG_ISFP_003_L4", "ISFP", "Creator_C", "Playful-Streetwear", "oversized", "Hoodie falling well below the waist", "Postmodern / Memphis Design / Y2K Loft"),
    ("IMG_ISFP_005_L1", "ISFP", "Creator_F", "Deconstructed", "oversized", "Extremely large grey zip-up hoodie", "Brutalism / Deconstructivism Interior"),
    ("IMG_ISFP_005_L2", "ISFP", "Creator_F", "Deconstructed", "layered", "Multiple overlapping shirts of different lengths", "Brutalism / Deconstructivism Interior"),
    ("IMG_ISFP_005_L3", "ISFP", "Creator_F", "Deconstructed", "dark_palette", "Washed black and faded grey color scheme", "Brutalism / Deconstructivism Interior"),
    ("IMG_ISFP_005_L4", "ISFP", "Creator_F", "Deconstructed", "edgy", "Silver hardware rings attached to the pants", "Brutalism / Deconstructivism Interior"),
    # ESFJ
    ("IMG_ESFJ_001_L1", "ESFJ", "Creator_C", "Clean-Traditional", "knitwear", "Navy blue v-neck sweater", "Classic Contemporary / Hamptons Style"),
    ("IMG_ESFJ_001_L2", "ESFJ", "Creator_C", "Clean-Traditional", "tailored", "Straight-cut khaki trousers", "Classic Contemporary / Hamptons Style"),
    ("IMG_ESFJ_001_L3", "ESFJ", "Creator_C", "Clean-Traditional", "leather", "Polished brown leather loafers", "Classic Contemporary / Hamptons Style"),
    ("IMG_ESFJ_001_L4", "ESFJ", "Creator_C", "Clean-Traditional", "minimalist", "No visible logos or complex patterns", "Classic Contemporary / Hamptons Style"),
    ("IMG_ESFJ_002_L1", "ESFJ", "Creator_A", "Ethereal-Soft", "flowing", "Long white chiffon skirt blowing gently", "Soft Minimal / Cloud Zen"),
    ("IMG_ESFJ_002_L2", "ESFJ", "Creator_A", "Ethereal-Soft", "pastel", "Light baby blue cardigan", "Soft Minimal / Cloud Zen"),
    ("IMG_ESFJ_002_L3", "ESFJ", "Creator_A", "Ethereal-Soft", "delicate", "Thin silver bracelet on the wrist", "Soft Minimal / Cloud Zen"),
    ("IMG_ESFJ_002_L4", "ESFJ", "Creator_A", "Ethereal-Soft", "layered", "Lightweight shawl draped over the arms", "Soft Minimal / Cloud Zen"),
    ("IMG_ESFJ_003_L1", "ESFJ", "Creator_D", "Soft-Romantic", "floral", "Midi dress with soft pink rose prints", "French Cottage / Vintage Shabby Chic"),
    ("IMG_ESFJ_003_L2", "ESFJ", "Creator_D", "Soft-Romantic", "feminine", "Fitted bodice highlighting the waist", "French Cottage / Vintage Shabby Chic"),
    ("IMG_ESFJ_003_L3", "ESFJ", "Creator_D", "Soft-Romantic", "elegant", "Classic beige Mary Jane heels", "French Cottage / Vintage Shabby Chic"),
    ("IMG_ESFJ_003_L4", "ESFJ", "Creator_D", "Soft-Romantic", "delicate", "Thin gold necklace with a pearl drop", "French Cottage / Vintage Shabby Chic"),
    ("IMG_ESFJ_005_L1", "ESFJ", "Creator_C", "Normcore", "casual", "Plain white cotton t-shirt", "Warm Scandinavian / Basic Japandi"),
    ("IMG_ESFJ_005_L2", "ESFJ", "Creator_C", "Normcore", "denim", "Straight fit medium wash blue jeans", "Warm Scandinavian / Basic Japandi"),
    ("IMG_ESFJ_005_L3", "ESFJ", "Creator_C", "Normcore", "knitwear", "Basic grey crewneck sweater", "Warm Scandinavian / Basic Japandi"),
    ("IMG_ESFJ_005_L4", "ESFJ", "Creator_C", "Normcore", "clean_silhouette", "Standard fit not too tight or too baggy", "Warm Scandinavian / Basic Japandi"),
    # ENFJ
    ("IMG_ENFJ_001_L1", "ENFJ", "Creator_D", "Clean-Traditional", "tailored", "Well-fitted navy blazer", "Classic Contemporary / Hamptons Style"),
    ("IMG_ENFJ_001_L2", "ENFJ", "Creator_D", "Clean-Traditional", "formal", "Crisp white button-down shirt", "Classic Contemporary / Hamptons Style"),
    ("IMG_ENFJ_001_L3", "ENFJ", "Creator_D", "Clean-Traditional", "knitwear", "Thin grey sweater worn over the shoulders", "Classic Contemporary / Hamptons Style"),
    ("IMG_ENFJ_001_L4", "ENFJ", "Creator_D", "Clean-Traditional", "clean_silhouette", "Tapered trousers with front creases", "Classic Contemporary / Hamptons Style"),
    ("IMG_ENFJ_002_L1", "ENFJ", "Creator_C", "Ethereal-Soft", "flowing", "Long silk slip dress falling loosely", "Soft Minimal / Cloud Zen"),
    ("IMG_ENFJ_002_L2", "ENFJ", "Creator_C", "Ethereal-Soft", "feminine", "Hair tied up with a large silk ribbon", "Soft Minimal / Cloud Zen"),
    ("IMG_ENFJ_002_L3", "ENFJ", "Creator_C", "Ethereal-Soft", "delicate", "Very thin gold chain with small pendant", "Soft Minimal / Cloud Zen"),
    ("IMG_ENFJ_002_L4", "ENFJ", "Creator_C", "Ethereal-Soft", "neutral_color", "Overall warm off-white palette", "Soft Minimal / Cloud Zen"),
    ("IMG_ENFJ_003_L1", "ENFJ", "Creator_A", "Soft-Romantic", "romantic", "Blouse with lace trim and puffed sleeves", "French Cottage / Vintage Shabby Chic"),
    ("IMG_ENFJ_003_L2", "ENFJ", "Creator_A", "Soft-Romantic", "vintage", "Retro-inspired structured handbag", "French Cottage / Vintage Shabby Chic"),
    ("IMG_ENFJ_003_L3", "ENFJ", "Creator_A", "Soft-Romantic", "elegant", "Polished black ballet flats", "French Cottage / Vintage Shabby Chic"),
    ("IMG_ENFJ_003_L4", "ENFJ", "Creator_A", "Soft-Romantic", "pastel", "Soft blue blouse with white skirt", "French Cottage / Vintage Shabby Chic"),
    ("IMG_ENFJ_005_L1", "ENFJ", "Creator_B", "Clean-Traditional", "formal", "Light blue button-down shirt", "Classic Contemporary / Hamptons Style"),
    ("IMG_ENFJ_005_L2", "ENFJ", "Creator_B", "Clean-Traditional", "leather", "Brown leather belt with simple buckle", "Classic Contemporary / Hamptons Style"),
    ("IMG_ENFJ_005_L3", "ENFJ", "Creator_B", "Clean-Traditional", "neutral_color", "Grey blue and brown combination", "Classic Contemporary / Hamptons Style"),
    ("IMG_ENFJ_005_L4", "ENFJ", "Creator_B", "Clean-Traditional", "layered", "Navy trench coat worn open", "Classic Contemporary / Hamptons Style"),
    # ESFP
    ("IMG_ESFP_001_L1", "ESFP", "Creator_D", "Glam-Maximalist", "leather", "Black leather pants with visible metal zippers", "Hollywood Regency / Art Deco Maximalism"),
    ("IMG_ESFP_001_L2", "ESFP", "Creator_D", "Glam-Maximalist", "bold_pattern", "Leopard print top paired with a bright red jacket", "Hollywood Regency / Art Deco Maximalism"),
    ("IMG_ESFP_001_L3", "ESFP", "Creator_D", "Glam-Maximalist", "glamorous", "Sparkly sequin mini dress catching the light", "Hollywood Regency / Art Deco Maximalism"),
    ("IMG_ESFP_001_L4", "ESFP", "Creator_D", "Glam-Maximalist", "layered", "Multiple chunky gold chains stacked on the neck", "Hollywood Regency / Art Deco Maximalism"),
    ("IMG_ESFP_002_L1", "ESFP", "Creator_A", "High-Energy", "sporty", "Neon pink track pants with side stripes", "High-Tech Dynamic / Neon Pop Modern"),
    ("IMG_ESFP_002_L2", "ESFP", "Creator_A", "High-Energy", "streetwear", "Oversized basketball jersey with large graphic numbers", "High-Tech Dynamic / Neon Pop Modern"),
    ("IMG_ESFP_002_L3", "ESFP", "Creator_A", "High-Energy", "casual", "White athletic tube socks pulled up high", "High-Tech Dynamic / Neon Pop Modern"),
    ("IMG_ESFP_002_L4", "ESFP", "Creator_A", "High-Energy", "layered", "Nylon windbreaker worn open over a sports bra", "High-Tech Dynamic / Neon Pop Modern"),
    ("IMG_ESFP_003_L1", "ESFP", "Creator_C", "Playful-Streetwear", "streetwear", "Extremely baggy light wash skater jeans", "Postmodern / Memphis Design / Y2K Loft"),
    ("IMG_ESFP_003_L2", "ESFP", "Creator_C", "Playful-Streetwear", "colorful", "Bright orange oversized graphic hoodie", "Postmodern / Memphis Design / Y2K Loft"),
    ("IMG_ESFP_003_L3", "ESFP", "Creator_C", "Playful-Streetwear", "casual", "Chunky skate shoes with thick laces", "Postmodern / Memphis Design / Y2K Loft"),
    ("IMG_ESFP_003_L4", "ESFP", "Creator_C", "Playful-Streetwear", "edgy", "Thick silver chain hanging from the belt loop", "Postmodern / Memphis Design / Y2K Loft"),
    ("IMG_ESFP_004_L1", "ESFP", "Creator_D", "Glam-Maximalist", "leather", "Shiny patent leather trench coat in black", "Hollywood Regency / Art Deco Maximalism"),
    ("IMG_ESFP_004_L2", "ESFP", "Creator_D", "Glam-Maximalist", "bold_pattern", "Zebra print flared trousers", "Hollywood Regency / Art Deco Maximalism"),
    ("IMG_ESFP_004_L3", "ESFP", "Creator_D", "Glam-Maximalist", "edgy", "Platform leather stilettos", "Hollywood Regency / Art Deco Maximalism"),
    ("IMG_ESFP_004_L4", "ESFP", "Creator_D", "Glam-Maximalist", "layered", "Faux fur scarf draped over one shoulder", "Hollywood Regency / Art Deco Maximalism"),
    ("IMG_ESFP_005_L1", "ESFP", "Creator_C", "High-Energy", "sporty", "Tight bright blue biker shorts", "High-Tech Dynamic / Neon Pop Modern"),
    ("IMG_ESFP_005_L2", "ESFP", "Creator_C", "High-Energy", "colorful", "High contrast between blue shorts and neon green socks", "High-Tech Dynamic / Neon Pop Modern"),
    ("IMG_ESFP_005_L3", "ESFP", "Creator_C", "High-Energy", "casual", "Nylon crossbody sporty bag strapped tightly across chest", "High-Tech Dynamic / Neon Pop Modern"),
    ("IMG_ESFP_005_L4", "ESFP", "Creator_C", "High-Energy", "bold_pattern", "Loud typography graphic print on the back of the jacket", "High-Tech Dynamic / Neon Pop Modern"),
    # ENFP
    ("IMG_ENFP_001_L1", "ENFP", "Creator_D", "High-Energy", "sporty", "Bright green matching track suit", "High-Tech Dynamic / Neon Pop Modern"),
    ("IMG_ENFP_001_L2", "ENFP", "Creator_D", "High-Energy", "streetwear", "Large graphic t-shirt with neon text", "High-Tech Dynamic / Neon Pop Modern"),
    ("IMG_ENFP_001_L3", "ENFP", "Creator_D", "High-Energy", "casual", "Standard running shoes", "High-Tech Dynamic / Neon Pop Modern"),
    ("IMG_ENFP_001_L4", "ENFP", "Creator_D", "High-Energy", "layered", "Neon mesh athletic top worn over a black sports bra", "High-Tech Dynamic / Neon Pop Modern"),
    ("IMG_ENFP_002_L1", "ENFP", "Creator_A", "Playful-Streetwear", "streetwear", "Wide leg jeans with paint splatters", "Postmodern / Memphis Design / Y2K Loft"),
    ("IMG_ENFP_002_L2", "ENFP", "Creator_A", "Playful-Streetwear", "colorful", "Rainbow striped knit sweater vest", "Postmodern / Memphis Design / Y2K Loft"),
    ("IMG_ENFP_002_L3", "ENFP", "Creator_A", "Playful-Streetwear", "casual", "Chunky colorful sneakers", "Postmodern / Memphis Design / Y2K Loft"),
    ("IMG_ENFP_002_L4", "ENFP", "Creator_A", "Playful-Streetwear", "edgy", "Multiple colorful enamel pins covering the jacket lapel", "Postmodern / Memphis Design / Y2K Loft"),
    ("IMG_ENFP_003_L1", "ENFP", "Creator_B", "Glam-Maximalist", "leather", "Metallic silver faux leather pants", "Hollywood Regency / Art Deco Maximalism"),
    ("IMG_ENFP_003_L2", "ENFP", "Creator_B", "Glam-Maximalist", "bold_pattern", "Bright pink faux fur coat", "Hollywood Regency / Art Deco Maximalism"),
    ("IMG_ENFP_003_L3", "ENFP", "Creator_B", "Glam-Maximalist", "edgy", "Platform black leather boots", "Hollywood Regency / Art Deco Maximalism"),
    ("IMG_ENFP_003_L4", "ENFP", "Creator_B", "Glam-Maximalist", "layered", "Multiple layered rhinestone necklaces", "Hollywood Regency / Art Deco Maximalism"),
    ("IMG_ENFP_004_L1", "ENFP", "Creator_A", "High-Energy", "sporty", "Huge vintage hockey jersey worn as a dress", "High-Tech Dynamic / Neon Pop Modern"),
    ("IMG_ENFP_004_L2", "ENFP", "Creator_A", "High-Energy", "colorful", "Neon yellow knitted beanie", "High-Tech Dynamic / Neon Pop Modern"),
    ("IMG_ENFP_004_L3", "ENFP", "Creator_A", "High-Energy", "casual", "Canvas tote bag slung over the shoulder", "High-Tech Dynamic / Neon Pop Modern"),
    ("IMG_ENFP_004_L4", "ENFP", "Creator_A", "High-Energy", "bold_pattern", "Cheetah print athletic socks pulled up to the calf", "High-Tech Dynamic / Neon Pop Modern"),
    ("IMG_ENFP_005_L1", "ENFP", "Creator_B", "Playful-Streetwear", "streetwear", "Denim overalls with one strap intentionally unbuttoned", "Postmodern / Memphis Design / Y2K Loft"),
    ("IMG_ENFP_005_L2", "ENFP", "Creator_B", "Playful-Streetwear", "colorful", "Bright tie-dye hoodie in pastel colors", "Postmodern / Memphis Design / Y2K Loft"),
    ("IMG_ENFP_005_L3", "ENFP", "Creator_B", "Playful-Streetwear", "casual", "Large puffy skate shoes in a contrasting color", "Postmodern / Memphis Design / Y2K Loft"),
    ("IMG_ENFP_005_L4", "ENFP", "Creator_B", "Playful-Streetwear", "edgy", "Heavily stickered skateboard carried under the arm", "Postmodern / Memphis Design / Y2K Loft"),
]

# -----------------------------------------------------------------------------
# 2. AUDIT TABLE + INTEGRITY CHECKS (Unit of Analysis: Image-Level)
# -----------------------------------------------------------------------------
rows = []
for s_id, mbti, creator, vibe, attr, ev, room in raw_look_data:
    desc, kw = mbti_ref[mbti]
    rows.append(dict(source_id=s_id, mbti=mbti, creator_account=creator,
                     fashion_vibe=vibe, fashion_attribute=attr, evidence=ev,
                     room_style=room, mbti_official_summary=desc,
                     mbti_core_keywords=kw))
df_raw = pd.DataFrame(rows)
df_raw["image_id"] = df_raw["source_id"].str.replace(r"_L\d+$", "", regex=True)

assert len(df_raw) == 268, f"expected 268 rows, got {len(df_raw)}"
assert df_raw["image_id"].nunique() == 67, "expected 67 images"
assert (df_raw.groupby("image_id").size() == 4).all(), "every image must have 4 rows"
assert (df_raw.groupby("image_id")["fashion_vibe"].nunique() == 1).all(), "one vibe per image"
assert df_raw["mbti"].nunique() == 16 and df_raw["fashion_vibe"].nunique() == 12

df_raw.to_csv("mbti_fashion_coded_data_audit.csv", index=False, encoding="utf-8-sig")

df_img = (df_raw.groupby("image_id")
          .agg(mbti=("mbti", "first"), creator=("creator_account", "first"),
               fashion_vibe=("fashion_vibe", "first"), room_style=("room_style", "first"))
          .reset_index())
df_img.to_csv("mbti_fashion_image_level.csv", index=False, encoding="utf-8-sig")
vibe_to_room = df_img.drop_duplicates("fashion_vibe").set_index("fashion_vibe")["room_style"].to_dict()

# -----------------------------------------------------------------------------
# 3. NPMI WEIGHTS + BIPARTITE GRAPH CONSTRUCTION
# -----------------------------------------------------------------------------
def npmi_edges(d):
    n = len(d)
    co = d.groupby(["mbti", "fashion_vibe"]).size().rename("co").reset_index()
    p_m = d["mbti"].value_counts() / n
    p_v = d["fashion_vibe"].value_counts() / n
    p_xy = co["co"] / n
    pmi = np.log(p_xy / (co["mbti"].map(p_m) * co["fashion_vibe"].map(p_v)))
    co["npmi"] = pmi / (-np.log(p_xy))
    return co

def build_graph(d, min_co=1):
    e = npmi_edges(d)
    e = e[(e["co"] >= min_co) & (e["npmi"] > 0)]
    g = nx.Graph()
    for _, r in e.iterrows():
        g.add_node(r["mbti"], node_type="mbti")
        g.add_node(r["fashion_vibe"], node_type="vibe")
        g.add_edge(r["mbti"], r["fashion_vibe"], weight=r["npmi"], dist=1.0 / r["npmi"], co=int(r["co"]))
    return g, e

edge_all = npmi_edges(df_img)
B, pruned = build_graph(df_img, min_co=1)
mbti_nodes = sorted(n for n, a in B.nodes(data=True) if a["node_type"] == "mbti")
vibe_nodes = sorted(n for n, a in B.nodes(data=True) if a["node_type"] == "vibe")

# -----------------------------------------------------------------------------
# 4. CENTRALITY & ZERO-OVERLAP COUPLE BRIDGE LOAD (Distance = 1/NPMI)
# -----------------------------------------------------------------------------
def vibe_table(g):
    ms = sorted(n for n, a in g.nodes(data=True) if a["node_type"] == "mbti")
    vs = sorted(n for n, a in g.nodes(data=True) if a["node_type"] == "vibe")
    b_dist = nx.betweenness_centrality(g, weight="dist")
    b_unw = nx.betweenness_centrality(g)
    pairs = list(itertools.combinations(ms, 2))
    cover = {v: sum(1 for a, b in pairs if g.has_edge(a, v) and g.has_edge(b, v)) for v in vs}
    no_share = [(a, b) for a, b in pairs if not set(g[a]) & set(g[b])]
    load = Counter()
    for a, b in no_share:
        paths = list(nx.all_shortest_paths(g, a, b, weight="dist"))
        for pth in paths:
            for node in pth[1:-1]:
                if node in vs:
                    load[node] += 1 / len(paths)
    t = pd.DataFrame({
        "Fashion Vibe": vs,
        "Room Style": [vibe_to_room[v] for v in vs],
        "Degree": [g.degree(v) for v in vs],
        "Betweenness (dist=1/NPMI)": [round(b_dist[v], 4) for v in vs],
        "Betweenness (unweighted)": [round(b_unw[v], 4) for v in vs],
        "Pairs covered (of %d)" % len(pairs): [cover[v] for v in vs],
        "Bridge load (no-shared pairs)": [round(load[v], 1) for v in vs],
    })
    return t.sort_values("Betweenness (dist=1/NPMI)", ascending=False).reset_index(drop=True), pairs, no_share, b_dist

df_vibe_summary, all_pairs, no_share_pairs, bet_dict = vibe_table(B)
primary_bridge = df_vibe_summary.iloc[0]["Fashion Vibe"]
top_cover = df_vibe_summary.sort_values("Pairs covered (of 120)", ascending=False).iloc[0]["Fashion Vibe"]

print("=== VIBE SUMMARY METRICS ===")
print(df_vibe_summary.to_string(index=False))

# -----------------------------------------------------------------------------
# 5. ROBUSTNESS (LEAVE-ONE-CREATOR-OUT)
# -----------------------------------------------------------------------------
rob = []
for creator in sorted(df_img["creator"].unique()):
    sub = df_img[df_img["creator"] != creator]
    g_sub, _ = build_graph(sub, 1)
    t_sub, _, _, _ = vibe_table(g_sub)
    rob.append({
        "dropped creator": creator,
        "images left": len(sub),
        "top bridge (1/NPMI)": t_sub.iloc[0]["Fashion Vibe"],
        "top coverage vibe": t_sub.sort_values("Pairs covered (of 120)", ascending=False).iloc[0]["Fashion Vibe"],
        "CT coverage": int(t_sub[t_sub["Fashion Vibe"] == "Clean-Traditional"]["Pairs covered (of 120)"].values[0])
    })
df_rob = pd.DataFrame(rob)
print("\n=== ROBUSTNESS TABLE ===")
print(df_rob.to_string(index=False))

# -----------------------------------------------------------------------------
# 6. FIGURE 1: TOPOLOGY VISUALIZATION (ESTJ & INFP SHORTEST MEDIATION PATH)
# -----------------------------------------------------------------------------
fig, ax = plt.subplots(figsize=(15, 10.5), dpi=300)

pos_exact = {
    "Glam-Maximalist": np.array([2.4, 8.8]),
    "ESFP": np.array([3.3, 9.1]),
    "ENFP": np.array([3.3, 8.5]),
    "Playful-Streetwear": np.array([2.7, 7.6]),
    "ESTP": np.array([3.2, 7.2]),
    "High-Energy": np.array([4.4, 7.7]),
    "ENTP": np.array([3.9, 7.3]),
    "Deconstructed": np.array([4.0, 6.2]),
    "ISFP": np.array([2.8, 5.6]),
    "Earthy-Raw": np.array([2.9, 4.8]),
    "INFP": np.array([2.3, 4.5]),
    "Soft-Romantic": np.array([3.8, 5.2]),
    "Ethereal-Soft": np.array([4.1, 3.4]),
    "ISFJ": np.array([4.8, 2.9]),
    "Utilitarian": np.array([5.5, 6.8]),
    "ISTP": np.array([5.6, 5.7]),
    "ENFJ": np.array([4.9, 5.2]),
    "INFJ": np.array([4.9, 4.4]),
    "ESFJ": np.array([5.6, 3.3]),
    "ENTJ": np.array([6.5, 6.0]),
    "INTJ": np.array([6.9, 7.0]),
    "Dark-Perfectionist": np.array([7.0, 5.5]),
    "Sharp-Tailored": np.array([8.2, 6.5]),
    "ISTJ": np.array([8.4, 6.0]),
    "ESTJ": np.array([9.1, 5.8]),
    "Clean-Traditional": np.array([6.7, 4.7]),
    "Normcore": np.array([6.8, 3.9]),
    "INTP": np.array([7.8, 4.0])
}

case_path = nx.shortest_path(B, source="ESTJ", target="INFP", weight="dist")
path_edges = list(zip(case_path[:-1], case_path[1:]))
hl_edges_matched = []
for u, v in path_edges:
    if B.has_edge(u, v):
        hl_edges_matched.append((u, v))
    elif B.has_edge(v, u):
        hl_edges_matched.append((v, u))

bg_edges = [e for e in B.edges() if e not in hl_edges_matched and (e[1], e[0]) not in hl_edges_matched]

nx.draw_networkx_edges(
    B, pos_exact, ax=ax,
    edgelist=bg_edges,
    width=[max(1.2, B[u][v]["weight"] * 4.5) for u, v in bg_edges],
    edge_color="#94A3B8",
    alpha=0.45
)

nx.draw_networkx_edges(
    B, pos_exact, ax=ax,
    edgelist=hl_edges_matched,
    width=5.5,
    edge_color="#EA580C",
    alpha=0.95
)

nx.draw_networkx_nodes(
    B, pos_exact, ax=ax,
    nodelist=mbti_nodes,
    node_color="#1F3A60",
    node_shape="s",
    node_size=1150
)

other_vibes = [v for v in vibe_nodes if v != primary_bridge]
nx.draw_networkx_nodes(
    B, pos_exact, ax=ax,
    nodelist=other_vibes,
    node_color="#FDBA74",
    node_size=1000
)

nx.draw_networkx_nodes(
    B, pos_exact, ax=ax,
    nodelist=[primary_bridge],
    node_color="#EA580C",
    node_size=1400,
    edgecolors="#0F172A",
    linewidths=3.5
)

nx.draw_networkx_labels(
    B, pos_exact, ax=ax,
    labels={n: n for n in mbti_nodes},
    font_size=8.5,
    font_weight="bold",
    font_color="white",
    font_family="sans-serif"
)

vibe_label_pos = {k: (v[0], v[1] - 0.22) for k, v in pos_exact.items() if k in vibe_nodes}
for node_name, (lx, ly) in vibe_label_pos.items():
    ax.text(
        lx, ly, node_name,
        fontsize=8.5,
        fontweight="bold",
        color="#0F172A",
        fontfamily="sans-serif",
        ha="center",
        va="top",
        bbox=dict(boxstyle="round,pad=0.15", facecolor="white", edgecolor="none", alpha=0.85)
    )

callout_text = (
    "Shortest Mediation Path for ESTJ & INFP (Zero Overlap):\n"
    "ESTJ → Sharp-Tailored → ... → Earthy-Raw → INFP\n"
    "Optimal Bridge Sequence: Sharp-Tailored → Dark-Perfectionist → Earthy-Raw"
)

target_x, target_y = pos_exact["Dark-Perfectionist"][0], pos_exact["Dark-Perfectionist"][1]
ax.annotate(
    callout_text,
    xy=(target_x, target_y),
    xytext=(6.8, 8.6),
    fontsize=9.0,
    fontweight="bold",
    color="#C2410C",
    fontfamily="sans-serif",
    ha="center",
    arrowprops=dict(
        arrowstyle="->",
        color="#EA580C",
        lw=2.0,
        shrinkA=6,
        shrinkB=12,
        connectionstyle="arc3,rad=-0.15"
    ),
    bbox=dict(boxstyle="round,pad=0.4", facecolor="#FFF7ED", edgecolor="#FDBA74", lw=1.2, alpha=0.95)
)

ax.set_title("Figure 1. MBTI–Fashion Vibe Bipartite Network (NPMI-weighted)",
             fontsize=13, fontweight="bold", pad=25)

subtitle_text = (
    f"Super Hub: Clean-Traditional (Degree = 8, Coverage = 28/120 pairs) | "
    f"Top Bridge: {primary_bridge} (Betweenness = 0.3134, Bridge Load = 34.8/58 pairs)\n"
    f"■ MBTI (Navy Square)   ● Fashion Vibe (Orange Circle)   | Bold Border = Top Bridge   | Orange Path = ESTJ-INFP Shortest Path"
)
fig.text(0.5, 0.02, subtitle_text, ha="center", fontsize=9.5, color="#334155")

ax.set_xlim(1.5, 9.8)
ax.set_ylim(2.2, 9.8)
ax.axis("off")

plt.tight_layout()
fig.savefig("mbti_network_topology_labeled.png", dpi=300, bbox_inches="tight")
plt.show()
print("✓ บันทึกรูปภาพ mbti_network_topology_labeled.png สำเร็จ")

# -----------------------------------------------------------------------------
# 7. INTERACTIVE SUITE: COUPLE MATCHER & VIS-NETWORK HTML
# -----------------------------------------------------------------------------
print("\n" + "=" * 65)
print("💑 COUPLE MATCHER ENGINE (ZERO-OVERLAP SHORTEST PATH SUPPORT)")
print("=" * 65)

dropdown_m1 = widgets.Dropdown(options=mbti_nodes, value="ESTJ", description='MBTI 1:', layout=widgets.Layout(width='220px'))
dropdown_m2 = widgets.Dropdown(options=mbti_nodes, value="INFP", description='MBTI 2:', layout=widgets.Layout(width='220px'))
btn_analyze = widgets.Button(description='วิเคราะห์ความเข้ากันได้', button_style='info', icon='heart')
out_text = widgets.Output()

def analyze_couple_action(b):
    with out_text:
        clear_output()
        m1 = dropdown_m1.value
        m2 = dropdown_m2.value
        if m1 == m2:
            print("⚠️ กรุณาเลือก MBTI ที่ต่างกัน 2 ประเภท")
            return

        vibes_a = set(B.neighbors(m1))
        vibes_b = set(B.neighbors(m2))
        shared = vibes_a.intersection(vibes_b)

        print(f"💑 ผลวิเคราะห์ความเข้ากันได้: {m1} & {m2}\n" + "-" * 55)
        if shared:
            print(f"✨ พบ Fashion Vibe ร่วมกัน {len(shared)} สไตล์ (Direct Overlap)\n")
            total_avg = sum((B[m1][v]['weight'] + B[m2][v]['weight']) / 2 for v in shared) / len(shared)
            print(f"ความกลมกลืนเฉลี่ยของคู่: {total_avg:.3f} NPMI\n")
            for v in sorted(shared, key=lambda x: (B[m1][x]['weight'] + B[m2][x]['weight']) / 2, reverse=True):
                avg = (B[m1][v]['weight'] + B[m2][v]['weight']) / 2
                print(f"• {v} — NPMI เฉลี่ย: {avg:.3f} ➔ 🏠 Room Style: {vibe_to_room.get(v, 'N/A')}")
        else:
            print("⚠️ ไม่มี Fashion Vibe ที่ทับซ้อนกันโดยตรง (Zero Direct Overlap)\n")
            try:
                path = nx.shortest_path(B, source=m1, target=m2, weight='dist')
                path_vibes = [n for n in path if n in vibe_nodes]
                print(f"🔍 ลำดับสะพานเชื่อมทางอ้อมที่มีแรงต้านน้อยที่สุด (Shortest Path: {len(path)-1} hops):")
                print(" -> ".join(path) + "\n")
                print("🏠 สไตล์ห้องที่แนะนำให้ใช้ประนีประนอม (Intermediate Room Styles):")
                for v in path_vibes:
                    print(f"• {v} ➔ 【 {vibe_to_room.get(v, 'N/A')} 】")
            except nx.NetworkXNoPath:
                print("ไม่พบเส้นทางเชื่อมโยงในเครือข่าย")

btn_analyze.on_click(analyze_couple_action)
display.display(widgets.HBox([dropdown_m1, dropdown_m2, btn_analyze]))
display.display(out_text)

# ข้อมูล JSON สำหรับ Vis-Network
edge_map = {m: {v: float(B[m][v]['weight']) for v in B.neighbors(m)} for m in mbti_nodes}
nodes_data = []

for node in mbti_nodes:
    nodes_data.append({
        'id': node, 'label': node, 'group': 'mbti', 'shape': 'box', 'margin': 12,
        'color': {'background': '#1F3A60', 'border': '#0E1F36', 'highlight': {'background': '#F39C12', 'border': '#D35400'}},
        'font': {'color': '#FFFFFF', 'size': 18, 'face': 'sans-serif', 'bold': True}
    })

for node in vibe_nodes:
    bet = bet_dict.get(node, 0.0)
    is_top_bridge = (node == primary_bridge)
    nodes_data.append({
        'id': node, 'label': node, 'group': 'vibe', 'shape': 'dot', 'size': 26 + (bet * 70),
        'color': {'background': '#EA580C' if is_top_bridge else '#FDBA74', 'border': '#9A3412', 'highlight': {'background': '#27AE60', 'border': '#1E8449'}},
        'font': {'color': '#0F172A', 'size': 14, 'face': 'sans-serif', 'bold': True, 'strokeWidth': 2, 'strokeColor': '#FFFFFF'},
        'title': f"<b>{node}</b><br>สไตล์ห้อง: {vibe_to_room.get(node, 'Unknown')}<br>Betweenness (1/NPMI): {bet:.4f}"
    })

edges_data = []
for u, v, data in B.edges(data=True):
    edges_data.append({
        'id': f"{u}__{v}",
        'from': u,
        'to': v,
        'value': max(1.5, data['weight'] * 4.5),
        'title': f"NPMI: {data['weight']:.3f}",
        'color': {'color': '#CBD5E1', 'highlight': '#EA580C', 'opacity': 0.8}
    })

# ฝัง HTML & JavaScript โดยแยก Header ออกจากกล่องกราฟ และคุมสีเฉพาะเส้นคู่ที่เลือก
html_code = f"""
<div style="width: 100%; border: 1px solid #E2E8F0; border-radius: 8px; font-family: -apple-system, BlinkMacSystemFont, 'Segoe UI', Roboto, sans-serif; background-color: #FAFAFA;">
  <!-- Header Bar อยู่นอกพื้นที่กราฟ ไม่บังโหนดแน่นอน -->
  <div style="display: flex; justify-content: space-between; align-items: center; padding: 12px 20px; background: #FFFFFF; border-bottom: 2px solid #E2E8F0; border-top-left-radius: 8px; border-top-right-radius: 8px;">
    <div id="instruction-bar" style="font-size: 14.5px; font-weight: 600; color: #1F3A60;">
      👉 <b>โหมดวิเคราะห์คู่แต่งงาน:</b> กรุณาคลิกเลือกโหนด <b>MBTI คนที่ 1</b> บนผังกราฟ
    </div>
    <button onclick="resetAllViews()" style="background-color: #1F3A60; color: white; border: none; padding: 8px 16px; font-size: 13px; font-weight: bold; border-radius: 6px; cursor: pointer; transition: 0.2s;">
      🔄 Reset View / พอดีหน้าจอ
    </button>
  </div>
  <!-- กล่องแสดงกราฟ Vis-Network เต็มพื้นที่ -->
  <div id="network-container" style="width: 100%; height: 680px;"></div>
</div>

<script type="text/javascript" src="https://unpkg.com/vis-network/standalone/umd/vis-network.min.js"></script>
<script type="text/javascript">
  (function() {{
    var rawNodes = {json.dumps(nodes_data)};
    var rawEdges = {json.dumps(edges_data)};
    var nodes = new vis.DataSet(rawNodes);
    var edges = new vis.DataSet(rawEdges);
    var container = document.getElementById('network-container');

    var data = {{ nodes: nodes, edges: edges }};
    var options = {{
      physics: {{
        forceAtlas2Based: {{ gravitationalConstant: -140, centralGravity: 0.015, springLength: 200, springConstant: 0.08, damping: 0.95, avoidOverlap: 1 }},
        solver: 'forceAtlas2Based', stabilization: {{ iterations: 200 }}
      }},
      interaction: {{ hover: true, navigationButtons: true, selectConnectedEdges: false }}
    }};

    var network = new vis.Network(container, data, options);
    var edgeMap = {json.dumps(edge_map)};
    var mbtiList = {json.dumps(list(mbti_nodes))};
    var selectedMBTI = [];

    network.on("click", function(params) {{
      if (params.nodes.length > 0) {{
        var clicked = params.nodes[0];
        if (mbtiList.includes(clicked)) {{
          if (selectedMBTI.length === 0) {{
            selectedMBTI.push(clicked);
            document.getElementById("instruction-bar").innerHTML = "💑 เลือกแล้ว: <b>" + clicked + "</b> ➔ กรุณาคลิกเลือก <b>MBTI คนที่ 2</b>";
            highlightSingle(clicked);
          }} else if (selectedMBTI.length === 1) {{
            if (selectedMBTI[0] === clicked) {{
              resetSelection();
              return;
            }}
            selectedMBTI.push(clicked);
            analyzePair(selectedMBTI[0], selectedMBTI[1]);
          }} else {{
            selectedMBTI = [clicked];
            document.getElementById("instruction-bar").innerHTML = "💑 เลือกแล้ว: <b>" + clicked + "</b> ➔ กรุณาคลิกเลือก <b>MBTI คนที่ 2</b>";
            highlightSingle(clicked);
          }}
        }}
      }}
    }});

    function highlightSingle(m) {{
      restoreEdgeStyles();
      var nodeUpdates = rawNodes.map(function(n) {{
        if (n.id === m) {{
          return {{ id: n.id, color: {{ background: '#F39C12', border: '#D35400' }} }};
        }} else {{
          return {{ id: n.id, color: n.color }};
        }}
      }});
      nodes.update(nodeUpdates);
    }}

    function analyzePair(m1, m2) {{
      var v1 = edgeMap[m1] ? Object.keys(edgeMap[m1]) : [];
      var v2 = edgeMap[m2] ? Object.keys(edgeMap[m2]) : [];
      var shared = v1.filter(v => v2.includes(v));

      if (shared.length > 0) {{
        document.getElementById("instruction-bar").innerHTML = "✅ เปรียบเทียบ: <b>" + m1 + "</b> & <b>" + m2 + "</b> (พบจุดร่วมโดยตรง " + shared.length + " สไตล์: " + shared.join(", ") + ")";

        var targetEdgeKeys = [];
        shared.forEach(function(v) {{
          targetEdgeKeys.push(m1 + "__" + v);
          targetEdgeKeys.push(v + "__" + m1);
          targetEdgeKeys.push(m2 + "__" + v);
          targetEdgeKeys.push(v + "__" + m2);
        }});

        // ปรับเส้น: ส้มเฉพาะเส้นเชื่อมระหว่างคู่นี้กับ Vibe ที่เป็นจุดร่วม ส่วนเส้นอื่นจางลง
        var edgeUpdates = rawEdges.map(function(e) {{
          if (targetEdgeKeys.includes(e.id)) {{
            return {{ id: e.id, color: {{ color: '#EA580C', opacity: 1.0 }}, width: 5.5 }};
          }} else {{
            return {{ id: e.id, color: {{ color: '#E2E8F0', opacity: 0.15 }}, width: 1.0 }};
          }}
        }});
        edges.update(edgeUpdates);

        // ไฮไลต์สีโหนด
        var activeNodes = [m1, m2].concat(shared);
        var nodeUpdates = rawNodes.map(function(n) {{
          if (n.id === m1 || n.id === m2) {{
            return {{ id: n.id, color: {{ background: '#F39C12', border: '#D35400' }} }};
          }} else if (shared.includes(n.id)) {{
            return {{ id: n.id, color: {{ background: '#27AE60', border: '#1E8449' }} }};
          }} else {{
            return {{ id: n.id, color: {{ background: '#F1F5F9', border: '#CBD5E1' }}, font: {{ color: '#94A3B8' }} }};
          }}
        }});
        nodes.update(nodeUpdates);

      }} else {{
        document.getElementById("instruction-bar").innerHTML = "⚠️ <b>" + m1 + "</b> & <b>" + m2 + "</b> ไม่มีจุดร่วมโดยตรง (ระบบแนะนำใช้ Shortest Path Mediation)";
        restoreEdgeStyles();
        var nodeUpdates = rawNodes.map(function(n) {{
          if (n.id === m1 || n.id === m2) {{
            return {{ id: n.id, color: {{ background: '#F39C12', border: '#D35400' }} }};
          }} else {{
            return {{ id: n.id, color: n.color, font: n.font }};
          }}
        }});
        nodes.update(nodeUpdates);
      }}
    }}

    function restoreEdgeStyles() {{
      edges.update(rawEdges.map(function(e) {{
        return {{ id: e.id, color: e.color, width: e.value }};
      }}));
      nodes.update(rawNodes.map(function(n) {{
        return {{ id: n.id, color: n.color, font: n.font }};
      }}));
    }}

    window.resetSelection = function() {{
      selectedMBTI = [];
      document.getElementById("instruction-bar").innerHTML = "👉 <b>โหมดวิเคราะห์คู่แต่งงาน:</b> กรุณาคลิกเลือกโหนด <b>MBTI คนที่ 1</b> บนผังกราฟ";
      restoreEdgeStyles();
      network.unselectAll();
    }};

    window.resetAllViews = function() {{
      window.resetSelection();
      network.fit({{animation: {{duration: 700, easingFunction: 'easeInOutQuad'}}}});
    }};
  }})();
</script>
"""

with open('mbti_fashion_vibe_network.html', 'w', encoding='utf-8') as f:
    f.write(html_code)

display.display(display.HTML(html_code))
```
