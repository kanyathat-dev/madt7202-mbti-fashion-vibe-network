## Why a Bipartite Network?

The project uses MBTI categories as one node set and Fashion Vibe categories as the other. This structure makes it possible to inspect which Vibes are associated with which MBTI labels and to explore indirect paths between MBTI types that do not share a direct Vibe.

This is not a direct MBTI-to-MBTI network. A path between two MBTI nodes alternates through Fashion Vibe nodes. Any highlighted sequence of Vibes should therefore be described alongside the complete path returned by the code when exact path structure matters.

## Graph Construction and Reported Structure

- **Node set A:** 16 MBTI categories.
- **Node set B:** 12 Fashion Vibe categories.
- **Total nodes:** 28.
- **Possible cross-set links:** 16 × 12 = 192.
- **Edge weight:** image-level NPMI.
- **NPMI range:** −1 to 1.
- **Pruning rule:** retain observed edges with **NPMI > 0**.
- **Reported retained graph:** 51 edges out of 192 possible cross-set links.
- **Shortest-path distance:** **d = 1/NPMI** for retained positive-weight edges.

NPMI is an association weight; the inverse transformation is used as a distance for shortest-path analysis. Positive NPMI does not establish causality or statistical significance. A missing edge should not be interpreted as proof that a preference does not exist outside this sample.

## Network Measures

- **Degree:** number of direct links attached to a node.
- **MBTI-pair coverage:** number of unordered MBTI pairs covered by a Vibe under the project's implemented definition, reported relative to the 120 possible pairs among 16 MBTI types.
- **Betweenness centrality:** measures how often a node lies on shortest paths under the graph and distance definition used by the code.
- **Bridge load:** the accumulated fractional path-load assigned to a Fashion Vibe across all MBTI pairs with no shared direct Vibe. For each pair, the code splits a contribution equally across tied weighted shortest paths and adds it to every intermediate Vibe. This is not a count of unique pairs or a percentage.
- **Shortest-path mediation:** an indirect route connecting MBTI categories that do not share a direct Vibe.
- **Leave-one-creator-out sensitivity:** compares bridge rankings after excluding each creator in turn.

Metric definitions, weighting, normalization, and tie-handling should follow the executed notebook code. Report any mismatch between these definitions and the actual implementation before submitting the analysis.

## 3. NPMI network and computed metrics

The graph retains edges with positive NPMI and uses `distance = 1 / NPMI`. Bridge Load is reported in fractional path-load units, separately from the percentage of MBTI pairs with no direct shared Vibe.

```python
# 3. NPMI WEIGHTS, BIPARTITE GRAPH, CENTRALITY, PAIR COVERAGE, AND BRIDGE LOAD
ALL_MBTI = sorted(df_img["mbti"].unique())
ALL_VIBES = sorted(df_img["fashion_vibe"].unique())

def npmi_edges(data):
    """Return image-level MBTI x Fashion Vibe co-occurrence and NPMI weights."""
    n = len(data)
    co = data.groupby(["mbti", "fashion_vibe"]).size().rename("co").reset_index()
    p_m = data["mbti"].value_counts() / n
    p_v = data["fashion_vibe"].value_counts() / n
    p_xy = co["co"] / n
    pmi = np.log(p_xy / (co["mbti"].map(p_m) * co["fashion_vibe"].map(p_v)))
    co["npmi"] = pmi / (-np.log(p_xy))
    return co

def build_graph(data, min_co=1):
    """Build a bipartite graph; keep positive-NPMI edges with co-occurrence >= min_co."""
    edges = npmi_edges(data)
    edges = edges[(edges["co"] >= min_co) & (edges["npmi"] > 0)].copy()
    graph = nx.Graph()
    # Fix the node universe across sensitivity runs so centrality normalization is comparable.
    graph.add_nodes_from((m, {"node_type": "mbti"}) for m in ALL_MBTI)
    graph.add_nodes_from((v, {"node_type": "vibe"}) for v in ALL_VIBES)
    for _, row in edges.iterrows():
        graph.add_edge(
            row["mbti"], row["fashion_vibe"],
            weight=float(row["npmi"]),
            dist=float(1.0 / row["npmi"]),
            co=int(row["co"])
        )
    return graph, edges

def vibe_table(graph):
    """Score every Vibe; count fractional path-load only where a path exists."""
    mbti_nodes = ALL_MBTI
    vibe_nodes = ALL_VIBES
    weighted_bc = nx.betweenness_centrality(graph, weight="dist")
    unweighted_bc = nx.betweenness_centrality(graph)
    pairs = list(itertools.combinations(mbti_nodes, 2))

    coverage = {
        vibe: sum(
            1 for a, b in pairs
            if graph.has_edge(a, vibe) and graph.has_edge(b, vibe)
        )
        for vibe in vibe_nodes
    }
    no_share = [
        (a, b) for a, b in pairs
        if not (set(graph.neighbors(a)) & set(graph.neighbors(b)))
    ]
    bridge_load = Counter()
    unreachable_no_share_pairs = 0

    for a, b in no_share:
        try:
            paths = list(nx.all_shortest_paths(graph, a, b, weight="dist"))
        except nx.NetworkXNoPath:
            unreachable_no_share_pairs += 1
            continue
        if not paths:
            unreachable_no_share_pairs += 1
            continue
        contribution = 1.0 / len(paths)
        for path in paths:
            for node in path[1:-1]:
                if node in vibe_nodes:
                    bridge_load[node] += contribution

    table = pd.DataFrame({
        "Fashion Vibe": vibe_nodes,
        "Room Style": [vibe_to_room[v] for v in vibe_nodes],
        "Degree": [graph.degree(v) for v in vibe_nodes],
        "Betweenness (dist=1/NPMI)": [round(weighted_bc.get(v, 0.0), 4) for v in vibe_nodes],
        "Betweenness (unweighted)": [round(unweighted_bc.get(v, 0.0), 4) for v in vibe_nodes],
        "Pairs covered (of 120)": [coverage[v] for v in vibe_nodes],
        "Bridge Load (fractional path-load units)": [round(bridge_load[v], 1) for v in vibe_nodes],
    }).sort_values(
        ["Betweenness (dist=1/NPMI)", "Fashion Vibe"], ascending=[False, True]
    ).reset_index(drop=True)

    return table, pairs, no_share, weighted_bc, unreachable_no_share_pairs

B, pruned = build_graph(df_img, min_co=1)
df_vibe_summary, all_pairs, no_share_pairs, bet_dict, unreachable_pairs = vibe_table(B)
primary_bridge = df_vibe_summary.iloc[0]["Fashion Vibe"]
top_bridge_row = df_vibe_summary.loc[df_vibe_summary["Fashion Vibe"] == primary_bridge].iloc[0]
top_bridge_betweenness = float(top_bridge_row["Betweenness (dist=1/NPMI)"])
top_bridge_load = float(top_bridge_row["Bridge Load (fractional path-load units)"])
top_cover = df_vibe_summary.sort_values("Pairs covered (of 120)", ascending=False).iloc[0]["Fashion Vibe"]
no_overlap_pct = 100 * len(no_share_pairs) / len(all_pairs)
mbti_nodes = ALL_MBTI
vibe_nodes = ALL_VIBES

assert len(all_pairs) == 120
assert len(df_vibe_summary) == 12
assert len(B.nodes) == 28
assert len(pruned) == 51
assert len(no_share_pairs) == 58
assert unreachable_pairs == 0

df_vibe_summary.to_csv("mbti_vibe_network_metrics.csv", index=False, encoding="utf-8-sig")

print("NETWORK METRICS (freshly derived from CSV)")
print(f"Graph: {B.number_of_nodes()} nodes | {B.number_of_edges()} retained positive-NPMI edges")
print(f"Possible MBTI pairs: {len(all_pairs)} | no shared direct Vibe: {len(no_share_pairs)}/{len(all_pairs)} = {no_overlap_pct:.1f}%")
print(f"Unreachable no-overlap pairs in baseline graph: {unreachable_pairs}")
print(f"Top bridge: {primary_bridge} | weighted betweenness = {top_bridge_betweenness:.4f} | bridge load = {top_bridge_load:.1f} fractional path-load units")
print(f"Highest direct pair coverage: {top_cover}")
display.display(df_vibe_summary)
```

## 4. Sensitivity analysis and implications for the business decision

The analysis removes one creator at a time and varies the minimum observed MBTI–Vibe co-occurrence threshold. Interpret the ranking together with graph coverage and the number of unreachable pairs; a change in the top bridge is a reason to test multiple styles, not to select a new universal winner.

```python
# 4. ROBUSTNESS / SENSITIVITY ANALYSIS
# A) Leave one creator out at a time; B) change the minimum observed co-occurrence threshold.

robustness_rows = []
for creator in sorted(df_img["creator"].unique()):
    subset = df_img[df_img["creator"] != creator].copy()
    g_sub, e_sub = build_graph(subset, min_co=1)
    t_sub, pairs_sub, no_share_sub, bc_sub, unreachable_sub = vibe_table(g_sub)
    top_row = t_sub.iloc[0]
    robustness_rows.append({
        "Dropped creator": creator,
        "Images left": len(subset),
        "Edges": g_sub.number_of_edges(),
        "Top bridge": top_row["Fashion Vibe"],
        "Top bridge weighted BC": top_row["Betweenness (dist=1/NPMI)"],
        "No-direct-overlap pairs (of 120)": len(no_share_sub),
        "Unreachable no-overlap pairs": unreachable_sub,
    })

df_rob = pd.DataFrame(robustness_rows)
df_rob.to_csv("leave_one_creator_out_sensitivity.csv", index=False, encoding="utf-8-sig")
display.display(df_rob)

bridge_counts = df_rob["Top bridge"].value_counts()
print("LEAVE-ONE-CREATOR-OUT INTERPRETATION")
print(f"Playful-Streetwear stays top in {int(bridge_counts.get('Playful-Streetwear', 0))}/6 runs.")
print("The primary recommendation should be treated as a shortlist, not a universal rule, when deleting a creator changes the top bridge.")
print("Business implication: test Playful-Streetwear alongside the alternatives that lead the leave-one-out runs before using a single style as a default.")

threshold_rows = []
for min_co in [1, 2, 3]:
    g_thr, e_thr = build_graph(df_img, min_co=min_co)
    t_thr, pairs_thr, no_share_thr, bc_thr, unreachable_thr = vibe_table(g_thr)
    active_mbti = sum(g_thr.degree(m) > 0 for m in ALL_MBTI)
    active_vibes = sum(g_thr.degree(v) > 0 for v in ALL_VIBES)
    components = list(nx.connected_components(g_thr))
    largest_component = max((len(c) for c in components), default=0)
    top_thr = t_thr.iloc[0]
    threshold_rows.append({
        "Minimum co-occurrence": min_co,
        "Retained edges": g_thr.number_of_edges(),
        "Active MBTI nodes": active_mbti,
        "Active Vibe nodes": active_vibes,
        "Connected components (incl. isolates)": len(components),
        "Largest component (nodes)": largest_component,
        "No-direct-overlap pairs (of 120)": len(no_share_thr),
        "Unreachable no-overlap pairs": unreachable_thr,
        "Top bridge": top_thr["Fashion Vibe"],
        "Top bridge weighted BC": top_thr["Betweenness (dist=1/NPMI)"],
        "Top bridge path-load units": top_thr["Bridge Load (fractional path-load units)"],
    })

df_threshold_sensitivity = pd.DataFrame(threshold_rows)
df_threshold_sensitivity.to_csv("npmi_threshold_sensitivity.csv", index=False, encoding="utf-8-sig")
print("MINIMUM CO-OCCURRENCE THRESHOLD SENSITIVITY")
display.display(df_threshold_sensitivity)
print("Interpretation guardrail: a different top bridge under a stricter threshold is not automatically a better recommendation. Check active-node coverage and unreachable pairs; a fragmented graph weakens the basis for recommending an indirect bridge.")
```

```python
# 7. INTERACTIVE SUITE: COUPLE MATCHER & VIS-NETWORK HTML
# -----------------------------------------------------------------------------
print("\n" + "=" * 65)
print("💑 COUPLE MATCHER ENGINE (ZERO-OVERLAP SHORTEST PATH SUPPORT)")
print("=" * 65)

dropdown_m1 = widgets.Dropdown(options=mbti_nodes, value="ESTJ", description='MBTI 1:', layout=widgets.Layout(width='220px'))
dropdown_m2 = widgets.Dropdown(options=mbti_nodes, value="INFP", description='MBTI 2:', layout=widgets.Layout(width='220px'))
btn_analyze = widgets.Button(description='วิเคราะห์ความเข้ากันได้', button_style='info', icon='heart')
out_text = widgets.Output()

def analyze_couple_action(b):
    with out_text:
        clear_output()
        m1 = dropdown_m1.value
        m2 = dropdown_m2.value
        if m1 == m2:
            print("⚠️ กรุณาเลือก MBTI ที่ต่างกัน 2 ประเภท")
            return

        vibes_a = set(B.neighbors(m1))
        vibes_b = set(B.neighbors(m2))
        shared = vibes_a.intersection(vibes_b)

        print(f"💑 ผลวิเคราะห์ความเข้ากันได้: {m1} & {m2}\n" + "-" * 55)
        if shared:
            print(f"✨ พบ Fashion Vibe ร่วมกัน {len(shared)} สไตล์ (Direct Overlap)\n")
            total_avg = sum((B[m1][v]['weight'] + B[m2][v]['weight']) / 2 for v in shared) / len(shared)
            print(f"ความกลมกลืนเฉลี่ยของคู่: {total_avg:.3f} NPMI\n")
            for v in sorted(shared, key=lambda x: (B[m1][x]['weight'] + B[m2][x]['weight']) / 2, reverse=True):
                avg = (B[m1][v]['weight'] + B[m2][v]['weight']) / 2
                print(f"• {v} — NPMI เฉลี่ย: {avg:.3f} ➔ 🏠 Room Style: {vibe_to_room.get(v, 'N/A')}")
        else:
            print("⚠️ ไม่มี Fashion Vibe ที่ทับซ้อนกันโดยตรง (Zero Direct Overlap)\n")
            try:
                path = nx.shortest_path(B, source=m1, target=m2, weight='dist')
                path_vibes = [n for n in path if n in vibe_nodes]
                print(f"🔍 ลำดับสะพานเชื่อมทางอ้อมที่มีแรงต้านน้อยที่สุด (Shortest Path: {len(path)-1} hops):")
                print(" -> ".join(path) + "\n")
                print("🏠 สไตล์ห้องที่แนะนำให้ใช้ประนีประนอม (Intermediate Room Styles):")
                for v in path_vibes:
                    print(f"• {v} ➔ 【 {vibe_to_room.get(v, 'N/A')} 】")
            except nx.NetworkXNoPath:
                print("ไม่พบเส้นทางเชื่อมโยงในเครือข่าย")

btn_analyze.on_click(analyze_couple_action)
display.display(widgets.HBox([dropdown_m1, dropdown_m2, btn_analyze]))
display.display(out_text)

# ข้อมูล JSON สำหรับ Vis-Network
edge_map = {m: {v: float(B[m][v]['weight']) for v in B.neighbors(m)} for m in mbti_nodes}
nodes_data = []

for node in mbti_nodes:
    nodes_data.append({
        'id': node, 'label': node, 'group': 'mbti', 'shape': 'box', 'margin': 12,
        'color': {'background': '#1F3A60', 'border': '#0E1F36', 'highlight': {'background': '#F39C12', 'border': '#D35400'}},
        'font': {'color': '#FFFFFF', 'size': 18, 'face': 'sans-serif', 'bold': True}
    })

for node in vibe_nodes:
    bet = bet_dict.get(node, 0.0)
    is_top_bridge = (node == primary_bridge)
    nodes_data.append({
        'id': node, 'label': node, 'group': 'vibe', 'shape': 'dot', 'size': 26 + (bet * 70),
        'color': {'background': '#EA580C' if is_top_bridge else '#FDBA74', 'border': '#9A3412', 'highlight': {'background': '#27AE60', 'border': '#1E8449'}},
        'font': {'color': '#0F172A', 'size': 14, 'face': 'sans-serif', 'bold': True, 'strokeWidth': 2, 'strokeColor': '#FFFFFF'},
        'title': f"<b>{node}</b><br>สไตล์ห้อง: {vibe_to_room.get(node, 'Unknown')}<br>Betweenness (1/NPMI): {bet:.4f}"
    })

edges_data = []
for u, v, data in B.edges(data=True):
    edges_data.append({
        'id': f"{u}__{v}",
        'from': u,
        'to': v,
        'value': max(1.5, data['weight'] * 4.5),
        'title': f"NPMI: {data['weight']:.3f}",
        'color': {'color': '#CBD5E1', 'highlight': '#EA580C', 'opacity': 0.8}
    })

# ฝัง HTML & JavaScript โดยแยก Header ออกจากกล่องกราฟ และคุมสีเฉพาะเส้นคู่ที่เลือก
html_code = f"""
<div style="width: 100%; border: 1px solid #E2E8F0; border-radius: 8px; font-family: -apple-system, BlinkMacSystemFont, 'Segoe UI', Roboto, sans-serif; background-color: #FAFAFA;">
  <!-- Header Bar อยู่นอกพื้นที่กราฟ ไม่บังโหนดแน่นอน -->
  <div style="display: flex; justify-content: space-between; align-items: center; padding: 12px 20px; background: #FFFFFF; border-bottom: 2px solid #E2E8F0; border-top-left-radius: 8px; border-top-right-radius: 8px;">
    <div id="instruction-bar" style="font-size: 14.5px; font-weight: 600; color: #1F3A60;">
      👉 <b>โหมดวิเคราะห์คู่แต่งงาน:</b> กรุณาคลิกเลือกโหนด <b>MBTI คนที่ 1</b> บนผังกราฟ
    </div>
    <button onclick="resetAllViews()" style="background-color: #1F3A60; color: white; border: none; padding: 8px 16px; font-size: 13px; font-weight: bold; border-radius: 6px; cursor: pointer; transition: 0.2s;">
      🔄 Reset View / พอดีหน้าจอ
    </button>
  </div>
  <!-- กล่องแสดงกราฟ Vis-Network เต็มพื้นที่ -->
  <div id="network-container" style="width: 100%; height: 680px;"></div>
</div>

<script type="text/javascript" src="https://unpkg.com/vis-network/standalone/umd/vis-network.min.js"></script>
<script type="text/javascript">
  (function() {{
    var rawNodes = {json.dumps(nodes_data)};
    var rawEdges = {json.dumps(edges_data)};
    var nodes = new vis.DataSet(rawNodes);
    var edges = new vis.DataSet(rawEdges);
    var container = document.getElementById('network-container');

    var data = {{ nodes: nodes, edges: edges }};
    var options = {{
      physics: {{
        forceAtlas2Based: {{ gravitationalConstant: -140, centralGravity: 0.015, springLength: 200, springConstant: 0.08, damping: 0.95, avoidOverlap: 1 }},
        solver: 'forceAtlas2Based', stabilization: {{ iterations: 200 }}
      }},
      interaction: {{ hover: true, navigationButtons: true, selectConnectedEdges: false }}
    }};

    var network = new vis.Network(container, data, options);
    var edgeMap = {json.dumps(edge_map)};
    var mbtiList = {json.dumps(list(mbti_nodes))};
    var selectedMBTI = [];

    network.on("click", function(params) {{
      if (params.nodes.length > 0) {{
        var clicked = params.nodes[0];
        if (mbtiList.includes(clicked)) {{
          if (selectedMBTI.length === 0) {{
            selectedMBTI.push(clicked);
            document.getElementById("instruction-bar").innerHTML = "💑 เลือกแล้ว: <b>" + clicked + "</b> ➔ กรุณาคลิกเลือก <b>MBTI คนที่ 2</b>";
            highlightSingle(clicked);
          }} else if (selectedMBTI.length === 1) {{
            if (selectedMBTI[0] === clicked) {{
              resetSelection();
              return;
            }}
            selectedMBTI.push(clicked);
            analyzePair(selectedMBTI[0], selectedMBTI[1]);
          }} else {{
            selectedMBTI = [clicked];
            document.getElementById("instruction-bar").innerHTML = "💑 เลือกแล้ว: <b>" + clicked + "</b> ➔ กรุณาคลิกเลือก <b>MBTI คนที่ 2</b>";
            highlightSingle(clicked);
          }}
        }}
      }}
    }});

    function highlightSingle(m) {{
      restoreEdgeStyles();
      var nodeUpdates = rawNodes.map(function(n) {{
        if (n.id === m) {{
          return {{ id: n.id, color: {{ background: '#F39C12', border: '#D35400' }} }};
        }} else {{
          return {{ id: n.id, color: n.color }};
        }}
      }});
      nodes.update(nodeUpdates);
    }}

    function analyzePair(m1, m2) {{
      var v1 = edgeMap[m1] ? Object.keys(edgeMap[m1]) : [];
      var v2 = edgeMap[m2] ? Object.keys(edgeMap[m2]) : [];
      var shared = v1.filter(v => v2.includes(v));

      if (shared.length > 0) {{
        document.getElementById("instruction-bar").innerHTML = "✅ เปรียบเทียบ: <b>" + m1 + "</b> & <b>" + m2 + "</b> (พบจุดร่วมโดยตรง " + shared.length + " สไตล์: " + shared.join(", ") + ")";

        var targetEdgeKeys = [];
        shared.forEach(function(v) {{
          targetEdgeKeys.push(m1 + "__" + v);
          targetEdgeKeys.push(v + "__" + m1);
          targetEdgeKeys.push(m2 + "__" + v);
          targetEdgeKeys.push(v + "__" + m2);
        }});

        // ปรับเส้น: ส้มเฉพาะเส้นเชื่อมระหว่างคู่นี้กับ Vibe ที่เป็นจุดร่วม ส่วนเส้นอื่นจางลง
        var edgeUpdates = rawEdges.map(function(e) {{
          if (targetEdgeKeys.includes(e.id)) {{
            return {{ id: e.id, color: {{ color: '#EA580C', opacity: 1.0 }}, width: 5.5 }};
          }} else {{
            return {{ id: e.id, color: {{ color: '#E2E8F0', opacity: 0.15 }}, width: 1.0 }};
          }}
        }});
        edges.update(edgeUpdates);

        // ไฮไลต์สีโหนด
        var activeNodes = [m1, m2].concat(shared);
        var nodeUpdates = rawNodes.map(function(n) {{
          if (n.id === m1 || n.id === m2) {{
            return {{ id: n.id, color: {{ background: '#F39C12', border: '#D35400' }} }};
          }} else if (shared.includes(n.id)) {{
            return {{ id: n.id, color: {{ background: '#27AE60', border: '#1E8449' }} }};
          }} else {{
            return {{ id: n.id, color: {{ background: '#F1F5F9', border: '#CBD5E1' }}, font: {{ color: '#94A3B8' }} }};
          }}
        }});
        nodes.update(nodeUpdates);

      }} else {{
        document.getElementById("instruction-bar").innerHTML = "⚠️ <b>" + m1 + "</b> & <b>" + m2 + "</b> ไม่มีจุดร่วมโดยตรง (ระบบแนะนำใช้ Shortest Path Mediation)";
        restoreEdgeStyles();
        var nodeUpdates = rawNodes.map(function(n) {{
          if (n.id === m1 || n.id === m2) {{
            return {{ id: n.id, color: {{ background: '#F39C12', border: '#D35400' }} }};
          }} else {{
            return {{ id: n.id, color: n.color, font: n.font }};
          }}
        }});
        nodes.update(nodeUpdates);
      }}
    }}

    function restoreEdgeStyles() {{
      edges.update(rawEdges.map(function(e) {{
        return {{ id: e.id, color: e.color, width: e.value }};
      }}));
      nodes.update(rawNodes.map(function(n) {{
        return {{ id: n.id, color: n.color, font: n.font }};
      }}));
    }}

    window.resetSelection = function() {{
      selectedMBTI = [];
      document.getElementById("instruction-bar").innerHTML = "👉 <b>โหมดวิเคราะห์คู่แต่งงาน:</b> กรุณาคลิกเลือกโหนด <b>MBTI คนที่ 1</b> บนผังกราฟ";
      restoreEdgeStyles();
      network.unselectAll();
    }};

    window.resetAllViews = function() {{
      window.resetSelection();
      network.fit({{animation: {{duration: 700, easingFunction: 'easeInOutQuad'}}}});
    }};
  }})();
</script>
"""

with open('mbti_fashion_vibe_network.html', 'w', encoding='utf-8') as f:
    f.write(html_code)

display.display(display.HTML(html_code))
```
